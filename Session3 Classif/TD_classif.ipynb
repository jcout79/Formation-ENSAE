```python
import pandas as pd
import numpy as np
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt
```

## K-means

```python
don = pd.read_table("donclassif.txt", header=0, sep=";")
don.head()
```

```python
# standardiser les variables
X = StandardScaler()
scaled_don = X.fit_transform(don)
scaled_don
```

```python
# lancer k-means k=2
kmeans2 = KMeans(n_clusters=2, random_state=0, n_init="auto")
kmeans2.fit(scaled_don)
```

```python
# lancer k-means k=3
kmeans3 = KMeans(n_clusters=3, random_state=0, n_init="auto")
kmeans3.fit(scaled_don)
```

```python
# récupérer les affectations de classes = étiquettes des clusters
print(kmeans2.labels_)
print(kmeans3.labels_)
```

```python
# Coordonnées des centres des clusters
print(kmeans2.cluster_centers_) 
print(kmeans3.cluster_centers_) 
```

```python
# visualisation des clusters
plt.scatter(scaled_don[:, 0], scaled_don[:, 1], c=kmeans3.labels_, cmap='viridis')
plt.scatter(kmeans3.cluster_centers_[:, 0], kmeans3.cluster_centers_[:, 1], s=300, c='red', marker='x')
plt.show()
```
