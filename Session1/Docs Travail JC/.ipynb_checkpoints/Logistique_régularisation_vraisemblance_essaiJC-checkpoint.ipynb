```python
# Définition des matrices X et Y, avec dichotomisation des X qualitatifs
nomsvar = list(SAh.columns.difference(["chd"]))
formule = "~ 1 +" + "+".join(nomsvar)
dsX = dmatrix(formule, data=SAh)
X = dmatrix(formule, data=SAh, return_type="dataframe").\
    iloc[:,1:].to_numpy()
Y = SAh["chd"].to_numpy()
```

```python
# Centrer/réduire les X pour les régressions pénalisées
scalerX = StandardScaler().fit(X)   # StandardScaler().fit(X) : Calcule la moyenne et l'écart-type de chaque colonne de la matrice X (les features) pour les standardiser (centrer-réduire).
Xcr= scalerX.transform(X)  # scalerX.transform(X) : Applique la standardisation à X, ce qui donne Xcr (une matrice où chaque colonne a une moyenne de 0 et un écart-type de 1).

# Calcul des facteurs d'échelle pour les pénalités
## Y - Y.mean() : Centre la variable cible Y (soustrait sa moyenne).
## Xcr.transpose().dot(...) : Calcule le produit matriciel entre Xcr (transposé) et (Y - Y.mean()), ce qui donne un vecteur de corrélations entre chaque feature et Y.
## np.abs(...).max() : Prend la valeur absolue maximale de ce vecteur (la corrélation la plus forte en valeur absolue).
## / X.shape[0] : Divise par le nombre d'observations (moyenne).
## Résultat : l0 est une estimation de l'importance maximale des features pour prédire Y, utilisée pour calibrer les valeurs des hyperparamètres de pénalisation.

l0 = np.abs(Xcr.transpose().dot((Y  - Y.mean()))).max()/X.shape[0]
# Création d'une grille de valeurs pour les hyperparamètres ridge et elasticnet à partir de la grille de lasso
## np.linspace(0, -4, 100) : Crée un vecteur de 100 valeurs régulièrement espacées entre 0 et -4 (échelle logarithmique).
## l0 * 10**llc : Applique une transformation exponentielle pour obtenir des valeurs de pénalisation (ll) couvrant plusieurs ordres de grandeur (de l0 à l0 * 10^-4).

llc = np.linspace(0,-4,100) 
ll = l0*10**llc

# Calcul des valeurs de C pour Lasso, Ridge et Elastic Net
Cs_lasso = 1/ 0.9/ X.shape[0] / (l0*10**(llc))
Cs_ridge = 1/ 0.9/ X.shape[0] / ((l0*10**(llc)) * 100)
Cs_enet =  1/ 0.9/ X.shape[0] / ((l0*10**(llc)) / 0.5)
```

Création des listes de coefficients pour chaque algo

```python
coefs_lasso = []
coefs_enet = []
coefs_ridge = []
```

Pour les 3 grilles associées à chaque méthodes, on lance lasso à partir de la reg logistique et géré les pénalités (lambda) + le chemin , puis ridge et elasticNet

```python
Xcr = StandardScaler().fit(X).transform(X)
for a, b, c in zip(Cs_lasso, Cs_ridge, Cs_enet) :
    ## lasso
    lasso = LogisticRegression(penalty="l1", C=a, solver="liblinear", warm_start=True).fit(Xcr, Y)
    coefs_lasso.append(lasso.coef_[0])
    ## ridge
    ridge = LogisticRegression(penalty="l2", C=b, warm_start=True).fit(Xcr, Y)
    coefs_ridge.append(ridge.coef_[0])
    ## enet
    enet = LogisticRegression(penalty="elasticnet", C=c, solver="saga", l1_ratio=0.5, warm_start=True).fit(Xcr, Y)
    coefs_enet.append(enet.coef_[0])
```
