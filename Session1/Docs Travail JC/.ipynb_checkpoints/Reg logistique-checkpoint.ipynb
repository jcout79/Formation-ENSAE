# Régression logistique via statsmodel GLM

```python
# formule pour sélectionner toutes les variables X (sauf la dernière colone qui est Y
form ="chd ~ " + "+".join(SAh.columns[ :-1 ])

# loi Binomiale car Y binaire, sinon multinomiale si categorielle nominale, sinon poisson si comptage
mod = smf.glm(form, data=SAh, family=sm.families.Binomial()).fit()
mod.summary()
```

```python
# l'indicateur AIC est dans le modèle, il suffit de le récupérer
print("AIC", mod.aic)
```
