# définition du fichier de travail

```python
don = pd.read_csv("ozone_complet.txt", header = 0, sep = ";", index_col=0)
don.shape
# à modifier si on veut prendre une partie du fichier / sélectionner un certain nombre de lignes pour tester l'overfitting
# don = don.iloc[0:50, :]
don.rename(columns={"maxO3":"Y"},inplace=True)
Y = don["Y"].to_numpy()
```

```python
# retirer données Na
lesna = pd.isna(don)
indNA = lesna.apply(any, axis=1)
don = don[~indNA]
don.shape
```

```python
# nb de blocs pour validation croisée
nb=10
tmp = np.arange(don.shape[0]) %nb 
rng = np.random.default_rng(seed=1234)
bloc = rng.choice(tmp,size=don.shape[0],replace=False)
```

# création du df de résultats des différents algo

```python
# n lignes = nb d'individus
# colonne identification du bloc d'appartenance
# colonne des Y
# colonnes MCO, BIC, etc : 1 colonne pour les Y prédits vec chaque algo
PREV = pd.DataFrame({"bloc":bloc,"Y":don["Y"],"MCO":0.0,"BIC":0.0,"AIC":0.0,
                    "ridge":0.0,"lasso":0.0,"elast":0.0,
                    "pls":0.0,"pcr":0.0 
                    })
```

# recodage des var quali en disjonctif

```python
nomsvar = list(don.columns.difference(["Y"]))
#design matrix : recodage de variables quali 
formule = "~" + "+".join(nomsvar)
dsX = dmatrix(formule,don)
### retire l'indicatrice des X ??? choisit de retirer quelle modalité de référence ??
X = np.asarray(dsX)[:,1:]
### features engineering : on ajoute les termes carrés et cubiques
# XX = np.concatenate((X, X**2, X**3), axis = 1)
### chercher comment intégrer les interactions entre les X
### Ajout des interactions (produits deux à deux des colonnes de X) p.259
### cf code prof
# X = XX
Y = don["Y"].to_numpy()
```

# définition des modèles de régression

```python
### kfregul: même kf pour tous les algos, permet de comparer les algos sur les mêmes blocs de validation croisée
kfregul = KFold(n_splits=10, shuffle=True, random_state=0) 
### kfaxes : ? sans doute pour la PLS et l'ACP 
### nbaxes : def du nb d'axes à conserver pour la PLS et l'ACP
kfaxes = KFold(n_splits=4, shuffle=True, random_state=0)
nbaxes = 20
# instanciation steps
cr = StandardScaler()
lassocv = LassoCV(cv=kfregul, n_jobs=3,max_iter=1000)
enetcv = ElasticNetCV(cv=kfregul, n_jobs=3,max_iter=1000)
# instanciation pipeline
pipe_lassocv = Pipeline(steps=[("cr", cr), ("lassocv", lassocv)])
pipe_enetcv = Pipeline(steps=[("cr", cr), ("enetcv", enetcv)])
## ridge : path
etape_lasso = pipe_lassocv.named_steps["lassocv"]
# instanciations
ridge = Ridge()
pipe_ridge = Pipeline(steps=[("cr", cr), ("ridge", ridge)])
acp = PCA()
reg = LinearRegression()
pipe_pcr = Pipeline(steps=[("cr", cr), ("acp", acp), ("reg", reg)])
regpls = PLSRegression()
## grille composantes et decoupage VC
param_grid_pcr = { "acp__n_components" : list(range(1,nbaxes))}
param_grid_pls = { "n_components" : list(range(1,nbaxes))}
```

# modélisation via validation croisée avec nb blocs

```python
# pour chaque bloc, renvoie 2 listes : variables conservées pour les modèles AIC et BIC.
# ne renvoie pas pour les autres méthodes car ce ne sont pas des méthodes de sélection de variables

### A FAIRE POUR AMELIORER CE CODE :
### mentionner le nom de chaque algo au lieu de 0, 1, 2, etc
### lister les noms des variables au lieu de leur rang dans la matrice X
### pour lasso : afficher les coefficients des variables

### récupérer code ridge sans gridsearch => dans code du cours
### faire code sans gridsearch pour ref sur composantes ACP et pour PLS ?

for i in np.arange(nb):
    print(i)
    Xapp = X[bloc!=i,:]
    Xtest = X[bloc==i,:]
    Yapp = don[bloc!=i]["Y"]
    Ytest = don[bloc==i]["Y"]
    #### reg
    reg = LinearRegression()
    reg.fit(Xapp,Yapp)
    PREV.loc[PREV.bloc==i,"MCO"] = reg.predict(Xtest)
    ### bic
    inst_reg_bic = ols_step_sk.LinearRegressionSelectionFeatureIC(verbose=1,crit="bic")
    reg_bic = inst_reg_bic.fit(X=Xapp, y=Yapp)
    PREV.loc[PREV.bloc==i,"BIC"] = reg_bic.predict(Xtest)
    ### aic
    inst_reg_aic = ols_step_sk.LinearRegressionSelectionFeatureIC(verbose=1,crit="aic")
    reg_aic = inst_reg_aic.fit(X=Xapp, y=Yapp)
    PREV.loc[PREV.bloc==i,"AIC"] = reg_aic.predict(Xtest)
    ## lasso
    pipe_lassocv.fit(Xapp,Yapp)
    PREV.loc[PREV.bloc==i,"lasso"] = pipe_lassocv.predict(Xtest)
    ## elastic net
    pipe_enetcv.fit(Xapp,Yapp)
    PREV.loc[PREV.bloc==i,"elast"] = pipe_enetcv.predict(Xtest)    
    ## params lambda
    path_ridge = etape_lasso.alphas_ * 100    
    param_grid_ridge = {"ridge__alpha": path_ridge}
    ## GridSearchCV
    cv_ridge = GridSearchCV(pipe_ridge, param_grid_ridge, cv=kfregul, scoring = "neg_mean_squared_error", n_jobs=3).fit(Xapp, Yapp)
    PREV.loc[PREV.bloc==i,"ridge"] = cv_ridge.predict(Xtest)
    ## gridsearch instanciation et fit
    cv_pcr = GridSearchCV(pipe_pcr, param_grid_pcr, cv=kfaxes, scoring = "neg_mean_squared_error", n_jobs=3).fit(Xapp,Yapp)
    cv_pls = GridSearchCV(regpls, param_grid_pls, cv=kfaxes, scoring = "neg_mean_squared_error", n_jobs=3).fit(Xapp,Yapp)
    PREV.loc[PREV.bloc==i,"pcr"] = cv_pcr.predict(Xtest)
    PREV.loc[PREV.bloc==i,"pls"] = cv_pls.predict(Xtest)
```

```python
# prev donne les erreurs de prévisions de chaque algo 
prev = PREV.iloc[:,1:]
np.round((prev.sub(PREV.Y, axis=0)**2).mean(),2)
```
