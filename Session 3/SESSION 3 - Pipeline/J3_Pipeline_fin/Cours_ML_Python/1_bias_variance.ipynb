# Introduction : Biais, Variance, Bruit

## 🎯 1. Décomposition biais-variance

La **formule de l'erreur de prédiction** peut être exprimée ainsi :

$$
\text{Erreur de prédiction} = \underbrace{\text{Biais}^2}_{\text{Erreur d’approximation}} + \underbrace{\text{Variance}}_{\text{Erreur d’estimation}} + \underbrace{\text{Bruit}}_{\text{Erreur irréductible}}
$$

**Erreur de prédiction** : erreur sur de nouvelles données (non vues à l'entraînement), liée à la capacité du modèle à généraliser.


### Relation entre l'erreur de prédiction individuelle et la décomposition biais-variance:

L'erreur de prédiction individuelle est la différence entre la valeur réelle $y$ et la prédiction du modèle $\hat{y}$ :

$$
\text{Erreur de prédiction} = y - \hat{y}
$$

Les métriques comme la MAE (Mean Absolute Error) ou la MSE (Mean Squared Error) sont des **moyennes de ces erreurs sur l'ensemble des données** :

$$
\text{MAE} = \frac{1}{n} \sum_{i=1}^{n} |y_i - \hat{y}_i|
$$

$$
\text{MSE} = \frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2
$$


#### 💡 **Biais² : Erreur d'approximation**  
Le **biais** est lié au **modèle lui-même** et reflète l'erreur introduite lorsque le modèle est **trop simple** pour capturer la complexité des données.  
Il représente la différence entre la valeur moyenne des prédictions du modèle et la vraie valeur $y$.

##### **Formule :**  
$$
Biais^2 = [E(f_D(x)) - y]^2
$$  

- $f_D(x)$ : prédiction du modèle entraîné sur un échantillon d'entraînement $D$.  
- $E(f_D(x))$ : moyenne des prédictions sur plusieurs jeux de données d'entraînement $D$.

##### **Interprétation :**  
Un biais élevé signifie que le modèle **sous-estime** la relation réelle, généralement parce qu'il est **trop simple**.  
- **Sous-apprentissage (underfitting)** : modèle trop rigide (ex. : régression linéaire sur une relation réellement quadratique).  
- Des hypothèses fortes sur la structure des données peuvent augmenter le biais.

##### **Exemple concret :**  
Si l'on utilise une régression linéaire pour prédire la croissance d'une plante dans le temps alors que la relation est quadratique, la moyenne des prédictions du modèle reste éloignée des valeurs réellement observées.


### 💡 Variance d'un modèle

La variance mesure à quel point les prédictions d’un modèle changent lorsque l’on modifie légèrement le jeu d’entraînement.  
Elle reflète la sensibilité du modèle aux fluctuations des données **et dépend fortement de la complexité du modèle**.

En général :

- un modèle très flexible (arbre profond, polynôme de haut degré, réseau large) a une variance élevée ;
- un modèle régularisé ou simple a une variance faible.

On peut exprimer la variance comme la variabilité des prédictions du modèle entraîné sur différents échantillons d'entraînement :

$$
Variance = E[(f_D(x) - E(f_D(x)))^2]
$$

où :

- $f_D(x)$ : prédiction du modèle entraîné sur un échantillon d'entraînement $D$ ;
- la variance mesure à quel point ces prédictions s'écartent de leur valeur moyenne $E(f_D(x))$ ;
- l'espérance est prise sur différents échantillons $D$.

Ainsi, la variance ne dépend pas uniquement des données :  
elle est fortement influencée par la complexité du modèle et par le niveau de régularisation appliqué.

## 🎯 3. Identification du Biais et de la Variance : Erreur d'entraînement et Erreur de test

Pour diagnostiquer le **biais** et la **variance**, on compare systématiquement l’**erreur d’entraînement** et l’**erreur de test**.

- **Erreur d'entraînement** : obtenue lorsque le modèle prédit sur les données qui ont servi à l’entraîner.  
  Elle peut être faible (modèle flexible comme un arbre non contraint) ou élevée (modèle rigide comme une régression linéaire mal adaptée).

- **Erreur de test** : obtenue sur des données **jamais vues** par le modèle.  
  Elle mesure la **capacité de généralisation**.

### 📊 Tableau d’interprétation

| Erreur d'entraînement | Erreur de test | Interprétation                          | Solutions possibles                                      |
|----------------------|----------------|-------------------------------------------|----------------------------------------------------------|
| Faible               | Faible         | Modèle bien adapté                        | Rien à changer                                           |
| Faible               | Élevée         | Surapprentissage → **variance élevée**    | Simplifier le modèle, augmenter la régularisation, utiliser un ensemble (bagging, random forest) |
| Élevée               | Élevée         | Sous-apprentissage → **biais élevé**       | Modèle plus complexe, ajouter des variables, augmenter la taille du dataset |


### 🔧 **Réduire la variance :**  
- Simplifier le modèle (pruning : suppression des branches peu significatives).  
- Utiliser des techniques d'ensemble (Bagging, Boosting).  
- Utiliser la validation croisée pour évaluer la robustesse du modèle et détecter un éventuel surapprentissage.

### 🔧 **Réduire le biais :**  
- Utiliser des modèles plus complexes (réseaux neuronaux).  
- Ajouter des variables pertinentes.  
- Augmenter la taille des données d'entraînement.  

## 🎯 [Optionnel] 4. Exercice Biais, Variance et Bruit – Exemple simplifié avec 2 échantillons  

Cet exemple illustre concrètement la décomposition de l'**erreur quadratique moyenne (EQM)** en **biais²**, **variance** et **bruit** à l'aide d'un exemple simple avec **deux training sets**.  

---

#### 📝 Hypothèses  

- **Fonction cible** : $f(x) = \sin(x)$  
- **Point d'évaluation** : $x_0 = 1$  
- **Valeur cible** : $f(1) \approx 0.84$  
- **Bruit aléatoire** :  
  $$
  \varepsilon \sim \mathcal{N}(0, \sigma^2) \quad \text{avec} \quad \sigma^2 = 0.01
  $$  
- **Deux prédictions issues de deux training sets différents** :  
  $$
  \hat{f}_1(1) = 0.60, \quad \hat{f}_2(1) = 0.70
  $$  

---

#### 💡 Objectif : Calculer l'EQM  
L'EQM se décompose en trois parties :  
$$
\text{EQM} = \text{Biais}^2 + \text{Variance} + \text{Bruit}
$$  

> 🔎 **Question pour les étudiants :**  
> Calculer chaque composante en suivant les étapes ci-dessous.  


### 🧮 Étape 1 – Moyenne des prédictions  

La **moyenne des prédictions** est donnée par :  
$$
\mathbb{E}[\hat{f}(x_0)] = \frac{1}{n} \sum_{i=1}^{n} \hat{f}_i(x_0)
$$  

---

### 📝 Étape 2 – Calcul du biais²  

Le **biais²** est la différence entre la **moyenne des prédictions** et la **valeur cible** :  
$$
\text{Biais}^2 = \left( \mathbb{E}[\hat{f}(x_0)] - f(x_0) \right)^2
$$  

---

### 📊 Étape 3 – Calcul de la variance  

La **variance** mesure la **dispersion** des prédictions autour de leur moyenne :  
$$
\text{Variance} = \frac{1}{n} \sum_{i=1}^{n} \left(\hat{f}_i(x_0) - \mathbb{E}[\hat{f}(x_0)]\right)^2
$$  

---

### 🌊 Étape 4 – Bruit (incompressible)  

Le **bruit** est donné par la variance de l'erreur aléatoire :  
$$
\text{Bruit} = \sigma^2
$$  

---

### 🧩 Résultat final : Erreur quadratique moyenne totale (MSE)  

L'**MSE** est la somme du biais², de la variance et du bruit :  
$$
\text{MSE} = \text{Biais}^2 + \text{Variance} + \text{Bruit}
$$  

---

### 📌 À retenir  
> - **Biais** : Erreur systématique due à un modèle trop simple.  
> - **Variance** : Sensibilité aux variations des données d'entraînement.  
> - **Bruit** : Part de l'erreur irréductible, due aux phénomènes aléatoires.  
> - La MSE est la **somme** des trois composantes.  
 L'EQM est la **somme** des trois composantes.  


## 🎯 5. Correction Exercice Biais, Variance et Bruit – Exemple simplifié avec 2 échantillons - 

### 🧮 Étape 1 – Moyenne des prédictions  

Pour estimer la **moyenne des prédictions**, on utilise la formule :  
$$
\mathbb{E}[\hat{f}(1)] = \frac{\hat{f}_1(1) + \hat{f}_2(1)}{2}
$$  

🔢 **Calcul** :  
$$
\mathbb{E}[\hat{f}(1)] = \frac{0.60 + 0.70}{2} = 0.65
$$  

---

### 📝 Étape 2 – Calcul du biais²  

Le biais représente la différence entre la **moyenne des prédictions** et la **valeur cible** :  
$$
\text{Biais}^2 = \left( \mathbb{E}[\hat{f}(1)] - f(1) \right)^2
$$  

🔢 **Calcul** :  
$$
\text{Biais}^2 = (0.65 - 0.84)^2 = 0.0361
$$  

> 🔎 **Question pour les étudiants :**  
> Quelle serait la valeur du biais² si $\mathbb{E}[\hat{f}(1)] = 0.70$ ?  

---

### 📊 Étape 3 – Calcul de la variance  

La variance mesure la **dispersion des prédictions** autour de leur moyenne :  
$$
\text{Variance} = \frac{1}{n} \sum_{i=1}^{n} \left(\hat{f}_i(1) - \mathbb{E}[\hat{f}(1)]\right)^2
$$  

🔢 **Calcul** :  
$$
\text{Variance} = \frac{1}{2} \left[(0.60 - 0.65)^2 + (0.70 - 0.65)^2\right] = 0.0025
$$  

> 🔎 **Question pour les étudiants :**  
> Si l'une des prédictions avait été 0.80 au lieu de 0.70, quelle serait la nouvelle variance ?  

---

### 🌊 Étape 4 – Bruit (incompressible)  

Le bruit est la **partie irréductible** de l'erreur :  
$$
\text{Bruit} = \sigma^2
$$  

🔢 **Calcul** :  
$$
\text{Bruit} = 0.01
$$  

---

### 🧩 Résultat final : Erreur quadratique moyenne totale (MSE)  

Pour obtenir la **MSE**, on additionne les trois composantes :  
$$
\text{MSE} = \text{Biais}^2 + \text{Variance} + \text{Bruit}
$$  

🔢 **Calcul via la décomposition** :  
$$
\text{MSE} = 0.0361 + 0.0025 + 0.01 = \boxed{0.0486}
$$  

#### 💡 **Calcul direct de la MSE :**  
La MSE est définie comme :  
$$
\text{MSE} = \frac{1}{n} \sum_{i=1}^{n} (f(1) - \hat{f}_i(1))^2
$$  

🔢 **Calcul direct** :  
$$
\text{MSE} = \frac{1}{2} \left[(0.84 - 0.60)^2 + (0.84 - 0.70)^2\right]
$$  
$$
\text{MSE} = \frac{1}{2} \times [0.0576 + 0.0196] = \frac{1}{2} \times 0.0772 = 0.0386
$$  
En ajoutant le bruit :  
$$
\text{MSE totale} = 0.0386 + 0.01 = \boxed{0.0486}
$$  

---

### ✅ **Conclusion :**  
Les deux méthodes (calcul direct et décomposition biais-variance-bruit) donnent le **même résultat** pour la MSE.  

#### 💡 **Pourquoi faire la décomposition ?**  
Bien que la MSE puisse être calculée directement, la **décomposition biais-variance-bruit** permet de :  
1. **Identifier la source de l'erreur** : Est-ce que l'erreur provient du modèle lui-même (biais), de la variabilité des données (variance) ou d'un bruit incompressible ?  
2. **Adapter la complexité du modèle** :  
   - Si l'erreur est due au biais ➔ Choisir un modèle plus complexe.  
   - Si l'erreur est due à la variance ➔ Régulariser ou simplifier le modèle.  
3. **Améliorer la généralisation** : La décomposition aide à équiibrer le compromis entre **biais et variance**, ce qui est essentiel pour obtenir un modèle robuste.  


#### 💡 **Pourquoi faire la décomposition biais-variance-bruit ?**  
Bien que la MSE soit directement calculable, la **décomposition biais-variance-bruit** est utile pour :  

1. **Comprendre l'origine de l'erreur :**  
   - Identifier si l'erreur provient du **biais** (modèle trop simple), de la **variance** (modèle trop complexe) ou d'un **bruit incompressible**.  
   - Exemple : Une régression linéaire sur des données non linéaires a un **biais élevé**, tandis qu'un arbre profond a une **variance élevée**.  

2. **Ajuster la complexité du modèle :**  
   - Si l'erreur vient du biais ➔ Choisir un modèle plus complexe.  
   - Si l'erreur vient de la variance ➔ Régulariser ou simplifier le modèle.  

3. **Relier théorie et pratique :**  
   - La décomposition biais-variance permet de mieux comprendre les sources d'erreur d'un modèle.  
   - Cependant, en pratique, on utilise surtout les **erreurs d'entraînement et de test** car elles sont plus simples à calculer et interpréter.
