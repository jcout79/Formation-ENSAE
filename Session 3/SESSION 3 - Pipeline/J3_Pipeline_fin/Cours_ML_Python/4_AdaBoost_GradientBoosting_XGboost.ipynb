# I. AdaBoost

## 1. Qu'est-ce que AdaBoost ?

✅ **AdaBoost** signifie **Adaptive Boosting**.  
Méthode d'agrégation de modèles faibles : 
- Entraîne des modèles faibles successivement,
- À chaque itération, renforce l’attention sur les erreurs des modèles précédents,
- Et combine tous les modèles avec un vote pondéré pour obtenir une prédiction finale.
- AdaBoost utilise souvent des **arbres de décision** très simples, appelés stumps.

```
              Couleur < 0.65 ?
              /             \
           Oui               Non
         Poire              Pomme

```
  
### Qu'est-ce qu'un modèle faible ?
- Qui fait légèrement mieux que le hasard,
- Mais n'est pas très performant seul.
  
Exemples de modèles faibles :
| Modèle faible | Exemple concret | Pourquoi faible ? |
|:-------------:|:----------------:|:-----------------:|
| Arbre de décision | Arbre de profondeur 1 (stump) | Très simple |
| Régression logistique | Peu entraînée | Frontière grossière |
| kNN | k = 1 sur petit dataset | Sur-apprend au bruit |
| Règle simple | "Poids > 170g ➔ Pomme" | Trop basique |

### Comment fonctionne AdaBoost ?
À chaque étape :
- Le modèle prédit la classe directement (pas les erreurs ni les résidus),
- On ajuste les poids des observations :
   * Les erreurs deviennent plus importantes,
   * Les bonnes prédictions deviennent moins importantes.
- ✅ À la fin, la prédiction globale est un vote pondéré de tous les modèles faibles.

## 🧠 2. Algorithme d'AdaBoost (classification)  

Supposons que $y_i \in \{-1, +1\}$ soit la vraie classe, et que $\hat{y}_i = h_m(x_i)$ soit la **prédiction** du modèle faible $h_m$ à l'itération $m$.

### 📐 À chaque itération $m$ :

1. **Entraîner un modèle faible** $h_m(x)$ sur les données pondérées $\{(x_i, y_i, w_i)\}$.  
   
2. **Calculer l'erreur pondérée** (erreur moyenne pondérée des prédictions incorrectes) :  
   $$
   \text{Erreur}_m = \sum_i w_i \cdot \mathbf{1}(\hat{y}_i \neq y_i)
   $$  
   > ⚠️ **Remarque :** Il s'agit d'une **somme pondérée des erreurs**. Plus le poids $w_i$ est élevé, plus l'erreur de prédiction est pénalisée.

3. **Calculer la force du modèle** (coefficient d'importance) :  
   $$
   \alpha_m = \frac{1}{2} \ln \left( \frac{1 - \text{Erreur}_m}{\text{Erreur}_m} \right)
   $$  
   > 🔑 **Interprétation :** Plus l'erreur est faible, plus le coefficient est grand, renforçant ainsi les modèles performants.

4. **Fonction de perte exponentielle (Exponential Loss)** :  
   La fonction de perte utilisée par AdaBoost est l'**erreur exponentielle**, équivalente à la **Exponential Loss** :  
   $$
   \text{Loss}(y_i, \hat{y}_i) = e^{-\alpha_m \cdot y_i \cdot \hat{y}_i}
   $$  
   - Lorsqu'une prédiction est correcte ($y_i \cdot \hat{y}_i > 0$), la perte est **faible**.  
   - Lorsqu'une prédiction est incorrecte ($y_i \cdot \hat{y}_i < 0$), la perte est **élevée**.  
   - Cette fonction de perte permet de **pondérer plus fortement** les erreurs pour que les modèles suivants les **corrigent prioritairement**.  
   > 📝 **Remarque :** La fonction exponentielle punit de manière **non linéaire** les erreurs, amplifiant leur impact.  

5. **Mise à jour des poids des exemples** :  
   La mise à jour des poids se fait en utilisant la fonction de perte exponentielle :  
   $$
   w_{i, m+1} = w_{i, m} \times \text{Loss}(y_i, \hat{y}_i)
   $$  
   Ce qui revient à :  
   $$
   w_{i, m+1} = w_{i, m} \times e^{-\alpha_m \cdot y_i \cdot \hat{y}_i}
   $$  
   - Si $\hat{y}_i = y_i$ (bonne prédiction), le poids **diminue**.  
   - Si $\hat{y}_i \neq y_i$ (mauvaise prédiction), le poids **augmente**.  

6. **Normaliser les poids** pour que leur somme soit égale à 1 :  
   $$
   w_i \leftarrow \frac{w_i}{\sum_i w_i}
   $$  
   > ✅ **But :** Garantir que les poids forment une distribution de probabilité.

---

### 🧩 Prédiction finale (combinaison des modèles faibles) :

Pour prédire la classe finale, AdaBoost réalise une **combinaison pondérée des prédictions des modèles faibles** :

$$
H(x) = \text{sign} \left( \sum_{m=1}^{M} \alpha_m \cdot h_m(x) \right)
$$

- $H(x)$ est la **prédiction finale**.
- $\alpha_m$ est la **force du modèle** $m$.
- $h_m(x)$ est la **prédiction du modèle faible** $m$.
- $\text{sign}(.)$ permet de donner le **signe final** (+1 ou -1).

> 🌟 **Interprétation :** La prédiction finale est une **somme pondérée des prédictions des modèles faibles**. Les modèles ayant un coefficient $\alpha_m$ élevé influencent davantage la décision.

---

### 🔍 Clarification de la terminologie :
- **Poids de l'exemple ($w_i$)** : Importance donnée à chaque exemple.  
- **Erreur pondérée** : Moyenne des erreurs pondérées par $w_i$.  
- **Force du modèle ($\alpha_m$)** : Poids donné au modèle dans le vote final (combinaison des modèles faibles).  
- **Exponential Loss (perte exponentielle)** : Fonction de perte qui amplifie les erreurs, utilisée pour ajuster les poids.  
- **Mise à jour des poids ($w_{i, m+1}$)** : Nouveau poids basé sur l'erreur pondérée.  
- **Prédiction finale ($H(x)$)** : Somme pondérée des prédictions des modèles faibles, prise avec le signe.  

---

### ✅ **Pourquoi utiliser la Exponential Loss ?**  
La **Exponential Loss (erreur exponentielle)** dans AdaBoost est essentielle pour :  
- **Penaliser fortement** les erreurs des modèles faibles.  
- **Réduire l'impact** des prédictions correctes.  
- **Concentrer l'apprentissage** sur les exemples difficiles.  
- **Renforcer** les modèles qui réussissent à corriger les erreurs précédentes.  

La fonction exponentielle étant **croissante**, les erreurs lourdes sont pénalisées de manière **non linéaire**, ce qui pousse les modèles suivants à se concentrer sur ces erreurs.  


### 📏 **Fonction de perte d'AdaBoost : Exponential Loss**

La fonction de perte est un concept central en **apprentissage supervisé**. Elle permet de **quantifier l'erreur** d'un modèle en comparant les **prédictions aux vraies valeurs**.  
Dans le cas d'AdaBoost, on utilise la **perte exponentielle**, qui pèse les erreurs de manière **exponentielle**.  

#### 💡 **Pourquoi une fonction de perte ?**  
- Elle sert à **évaluer la performance** de chaque modèle faible.  
- Elle permet de **pondérer les erreurs** pour donner plus d'importance aux erreurs difficiles à corriger.  
- La mise à jour des poids dans AdaBoost est **directement liée à la fonction de perte**.  

---

#### 📐 **Formule de la perte exponentielle :**  
$$
\text{Loss}(y, \hat{y}) = e^{-y \cdot \hat{y}}
$$  

- Lorsqu'une prédiction est correcte ($y \cdot \hat{y} > 0$), la perte est **faible**.  
- Lorsqu'une prédiction est incorrecte ($y \cdot \hat{y} < 0$), la perte est **élevée**.  
- Plus l'erreur est grande, plus la **pénalité augmente exponentiellement**.  
- Cela signifie que les **erreurs importantes sont fortement pénalisées**, ce qui pousse le modèle suivant à **se concentrer sur les erreurs**.  

---

#### 🌟 **Pourquoi utiliser -1 et +1 pour $y$ et $\hat{y}$ ?**  
- En classification binaire avec AdaBoost, les classes sont généralement représentées par **-1** et **+1**.  
- Cela permet de **simplifier le calcul de la perte** :  
  - Si la prédiction est **correcte** ($y \cdot \hat{y} > 0$), le produit est **+1**, donc :  
    $$
    \text{Loss} = e^{-\alpha \times 1} = e^{-\alpha}
    $$  
    > La perte est **faible** (inférieure à 1).  
  - Si la prédiction est **incorrecte** ($y \cdot \hat{y} < 0$), le produit est **-1**, donc :  
    $$
    \text{Loss} = e^{-\alpha \times (-1)} = e^{\alpha}
    $$  
    > La perte est **élevée** (supérieure à 1).  

---

#### 📝 **Exemple intuitif :**  
- Si un modèle faible prédit **correctement** un point, la perte est **très faible** (proche de zéro).  
- Si un modèle faible se **trompe** complètement, la perte est **très élevée**, et le poids de ce point est **augmenté** pour attirer l'attention du modèle suivant.  
- Cela pousse le modèle suivant à **corriger prioritairement** les points mal classés lors de l'itération précédente.  

---

#### ✅ **Pourquoi cette approche est efficace ?**  
- En amplifiant les erreurs importantes, AdaBoost **force les modèles faibles suivants à se concentrer sur les points difficiles**.  
- Cela permet d'**améliorer progressivement la précision globale** en **corrigeant en priorité les erreurs les plus critiques**.  


# 🌟 Comprendre AdaBoost : Sélection de la meilleure coupure, itérations et rôle du random_state

## 1️⃣ **AdaBoost cherche la meilleure coupure en fonction de l'erreur pondérée**

AdaBoost est un algorithme de boosting qui construit une séquence de **modèles faibles** pour améliorer la précision des prédictions.  
Contrairement aux arbres de décision classiques qui utilisent des critères comme l'**indice de Gini** ou l'**entropie**, AdaBoost utilise la **réduction d'erreur pondérée** pour sélectionner la meilleure coupure.  

### 💡 Comment AdaBoost choisit-il la meilleure coupure ?
- À chaque **itération**, AdaBoost cherche la coupure qui **minimise l'erreur pondérée**.  
- L'erreur pondérée est calculée comme suit :  

$$
\epsilon = \sum_{i=1}^{N} w_i \cdot \mathbb{1}(y_i \neq h(x_i))
$$  

- $w_i$ : Poids de l'échantillon $i$  
- $\mathbb{1}(y_i \neq h(x_i))$ : 1 si l'échantillon est mal classé, 0 sinon


### 🔄 **Pourquoi utiliser l'erreur pondérée ?**
- AdaBoost met plus de **poids sur les erreurs précédentes** pour que le modèle suivant se concentre sur leur correction.  
- En augmentant les poids des erreurs, AdaBoost donne **plus d'importance aux échantillons mal classés** lors des itérations suivantes.

## 3️⃣ **Le rôle du `random_state` dans AdaBoost**

### 💡 **Pourquoi fixer `random_state` dans AdaBoost ?**  
- Fixer `random_state` garantit que l'algorithme produit **les mêmes résultats** à chaque exécution.  
- **AdaBoost est déterministe** si :  
  - Les **données sont identiques**.  
  - Le **nombre d'estimateurs** est fixé.  
  - Les **poids initiaux** sont les mêmes.  
- Cependant, dans des cas très spécifiques, comme des **coupures équivalentes** ou **égalités parfaites**, l'algorithme peut faire un **choix aléatoire** entre deux coupures optimales.  

### 🔍 **Impact du `random_state` dans AdaBoost**  
- Avec des **stumps (arbres de profondeur 1)** et des données bien séparées, le **random_state** n'a **généralement pas d'effet**.  
- Pour des modèles faibles plus complexes (arbres plus profonds), le `random_state` peut **influencer le choix de la coupure**.  
- En fixant `random_state`, on garantit que l'algorithme **produit toujours les mêmes modèles faibles**, assurant ainsi la **reproductibilité** des résultats.

- **Erreur :** Le modèle fait une erreur sur **C** et **E**.

### 2. **Calculer l'erreur pondérée :**  
$$
\text{Erreur}_2 = w_C + w_E = 0.1112 + 0.1112 = 0.2224
$$  

---

### 3. **Calculer la force du modèle :**  
$$
\alpha_2 = \frac{1}{2} \ln \left(\frac{1 - \text{Erreur}_2}{\text{Erreur}_2}\right) \approx 0.626
$$  

---

### 4. **Calculer la fonction de perte (Exponential Loss)** :  
- Pour les erreurs (C et E) :  
$$
\text{Loss}(error) = e^{\alpha_2} = e^{0.626} \approx 1.870
$$  
- Pour les bonnes prédictions :  
$$
\text{Loss}(correct) = e^{-\alpha_2} = e^{-0.626} \approx 0.535
$$

$$
w_{new} = 0.1112 \times 1.870 \approx 0.202

w_{new} = 0.3737 \times 0.535 \ -->approx 0.148
$$

### 6. **Normaliser les poids :**  
$$
w_i \leftarrow \frac{w_i}{\sum_i w_i}
$$  
- Somme des poids après normalisation : $1$

### ✅ **Conclusion :**  
AdaBoost combine des modèles faibles pour obtenir un modèle robuste.  
Même si chaque modèle est imparfait, leur combinaison pondérée permet d'améliorer la précision globale.  
La **fonction de perte exponentielle** joue un rôle essentiel pour **renforcer l'importance des erreurs**, permettant au modèle de **corriger ses faiblesses** au fil des itérations.





## 🚀 Pourquoi AdaBoost améliore la performance ?

✅ À chaque itération :  
- Les erreurs sont rendues **plus visibles** grâce à la **perte exponentielle**.  
- Les modèles suivants se concentrent sur ce que les précédents ont **raté**.  
- Les meilleurs modèles sont **plus écoutés** grâce à leur poids $\alpha_m$.  




## ⚠️ Limites d'AdaBoost  
- Sensible aux **outliers**.  
- Peut **surapprendre** (overfitting) si le modèle est trop complexe.  
- Fonction de coût fixe (**perte exponentielle**), ce qui peut être sous-optimal pour certaines données.  

Pour surmonter ces limitations, **Gradient Boosting** propose une approche plus flexible et robuste en utilisant des résidus, des fonctions de perte personnalisables et des techniques de régularisation.

- **But :** Réduire le biais  
- **Méthode :**  
  - Modèles entraînés **séquentiellement**

- **Exemples :** AdaBoost, Gradient Boosting  

---

- **Bagging :** indépendance des modèles → réduction de la variance  
- **Boosting :** dépendance des modèles → réduction du biais  
amélioration itérative des erreurs**.  


# II. Gradient Boosting

## 1. Qu'est-ce que le Gradient Boosting ?

- ✅ Gradient Boosting est une méthode qui :  
  - Entraîne des modèles faibles de manière **séquentielle**,  
  - À chaque étape, corrige l'erreur de prédiction précédente,  
  - En suivant la direction du **gradient d'une fonction de coût** chisie.  

---

## 2. Comment fonctionne Gradient Boosting ?

Le principe est :  
- **Initialisation du modèle** (par exemple, prédiction initiale basée sur la moyenne pour la régression ou sur la log odds pour la classification).  
- **Calcul des erreurs résiduelles** (différence entre la valeur réelle et la valeur prédite).  
- **Entraînement d'un modèle faible sur les résidus**.  
- Ces résidus représentent en réalité **le gradient de la fonction de perte** (d'où le nom "Gradient Boosting").  

> 🔁 À chaque itération, on calcule le **gradient de la fonction de coût**  
> et on entraîne un modèle faible pour **corriger dans la direction opposée** (descente de gradient).  

- **Correction progressive des prédictions**.  
- ✅ À chaque itération, la prédiction s'améliore progressivement.  

---

## 3. Fonction de perte dans le Gradient Boosting  

La fonction de perte guide l'optimisation dans le Gradient Boosting en indiquant **l'écart entre la prédiction et la valeur réelle**.  

### 🌟 Fonctions de perte typiques :  

- **Pour la régression** (erreur quadratique moyenne) :  
  $$
  L(y, \hat{y}) = \frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2
  $$  
  > 🔎 Cette fonction mesure l'**écart quadratique moyen** entre les valeurs réelles et prédites.  

- **Pour la classification binaire** (log-loss) :  
  $$
  L(y, \hat{p}) = - \left[ y \cdot \log(\hat{p}) + (1 - y) \cdot \log(1 - \hat{p}) \right]
  $$  
  > 🔎 Cette fonction pénalise fortement les **prédictions incorrectes** en utilisant une fonction logarithmique.  

---

### ✅ **Résumé :**  
Gradient Boosting utilise une suite de modèles faibles pour réduire l'erreur en suivant le **gradient de la fonction de perte**.  
L'initialisation dépend de la tâche :  
- Pour la régression : moyenne des cibles.  
- Pour la classification : log odds des classes.  


## 4. Algorithme du Gradient Boosting (classification binaire)

On cherche à prédire une variable cible $y \in \{0, 1\}$ en minimisant la **log-loss**.

---

### Étape 1 : Initialisation

On commence avec une prédiction constante basée sur les log-odds de la classe 1 :

$$
F_0(x) = \frac{1}{2} \ln \left( \frac{p}{1 - p} \right), \quad \text{où } p = \frac{1}{n} \sum_{i=1}^n y_i
$$

---

### Étape 2 : Pour chaque itération $m = 1, 2, ..., M$

#### a. Calcul de la fonction de perte (log-loss)  

$$
L(y, \hat{p}) = - \left[ y \cdot \log(\hat{p}) + (1 - y) \cdot \log(1 - \hat{p}) \right]
$$

#### b. Calcul des pseudo-résidus  

Les pseudo-résidus sont calculés comme le **négatif du gradient de la log-loss** :

$$
r_i = y_i - \hat{p}_i, \quad \text{où } \hat{p}_i = \frac{1}{1 + e^{-F_{m-1}(x_i)}}
$$

> 🔄 Ces pseudo-résidus représentent **la direction de correction** basée sur la fonction de perte.

#### c. Apprentissage du modèle faible $h_m(x)$  

On entraîne un modèle $h_m(x)$ (par exemple, un arbre de décision simple) pour **prédire les $r_i$**.

#### d. Mise à jour du modèle

$$
F_m(x) = F_{m-1}(x) + \gamma \cdot h_m(x)
$$

où# $\gamma$ est le **learning rate**.

---

## Prédiction finale

### a. Conversion en probabilité

$$
\hat{p}(x) = \frac{1}{1 + e^{-F_M(x)}}
$$

### b. Classe prédite

$$
\hat{y} =
\begin{cases}
1 & \text{si } \hat{p}(x) \geq 0.5 \\
0 & \text{sinon}
\end{cases}
$$

---

## ✅ Pourquoi intégrer la fonction de perte dans l'algorithme ?  
- La fonction de perte guide la **direction de la correction** à chaque étape.  
- Les pseudo-résidus sont calculés comme **le gradient négatif de la fonction de perte**.  
- Intégrer la fonction de perte permet de **corriger les erreurs de manière progressive** en se concentrant sur les erreurs les plus significatives.  


## 5. Algorithme du Gradient Boosting (régression)

Le Gradient Boosting pour la régression cherche à prédire une variable continue en minimisant l'erreur quadratique moyenne (**MSE**).

---

### Étape 1 : Initialisation

On commence avec une prédiction constante, par exemple la **moyenne des valeurs réelles** :

$$
F_0(x) = \frac{1}{n} \sum_{i=1}^{n} y_i
$$

---

### Étape 2 : Pour chaque itération $m = 1, 2, ..., M$

#### a. Calcul de la fonction de perte (MSE)  

La fonction de perte est l'**erreur quadratique moyenne** :

$$
L(y, \hat{y}) = \frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2
$$

> 🔄 **Rôle de la fonction de perte :**  
> La MSE permet de **mesurer l'écart quadratique moyen** entre les valeurs réelles et les prédictions.  
> Elle est utilisée pour **quantifier l'erreur** et orienter la correction.  

---

#### b. Calcul des résidus

Les résidus représentent le **gradient négatif de la fonction de perte (MSE)** :

$$
r_i = y_i - F_{m-1}(x_i)
$$

> 🔎 Les résidus représentent **l'écart entre la valeur réelle et la prédiction actuelle**.  
> Ce sont les valeurs sur lesquelles le **modèle faible** va s'entraîner pour corriger l'erreur.  

---

#### c. Apprentissage du modèle faible $h_m(x)$

On entraîne un modèle faible (par exemple, un **arbre de décision simple**) pour **prédire les résidus $r_i$**.  
Le but est de **trouver la fonction qui minimise la perte** sur les résidus.  

---

#### d. Mise à jour de la prédiction globale

La prédiction est mise à jour en ajoutant la contribution du modèle faible pondérée par le **learning rate ($\gamma$)** :

$$
F_m(x) = F_{m-1}(x) + \gamma \cdot h_m(x)
$$

> 🔑 **Pourquoi ajouter le modèle faible ?**  
> Le modèle faible $h_m(x)$ **corrige l'erreur** de la prédiction précédente.  
> Le learning rate $\gamma$ permet de **pondérer la correction** pour éviter un ajustement trop rapide.  

---

### ✅ **Pourquoi intégrer la fonction de perte dans l'algorithme ?**  
- La fonction de perte (MSE) est utilisée pour **quantifier l'erreur** et **orienter la correction**.  
- Les résidus sont calculés comme le **gradient de cette fonction de perte**.  
- L'intégration de la fonction de perte garantit que chaque itération **corrige progressivement l'erreur** en suivant la **direction optimale**.

# 🎯 Comprendre la différence entre Loss Function et Metric

Lorsqu'on entraîne un modèle de Machine Learning, il est crucial de distinguer :

| Terme | Rôle |
|------|------|
| **Loss function** (fonction de coût) | Sert à **optimiser l'entraînement** du modèle. C’est ce que l’algorithme essaie de **minimiser**. |
| **Metric** (métrique de performance) | Sert à **évaluer la qualité finale** de la prédiction. Elle est calculée **après** l'entraînement, souvent sur les données de test. |

## Exemple concret (classification binaire)

- Le modèle est entraîné pour **minimiser une loss** (ex : log-loss).
- Mais on **juge ses performances** avec une métrique comme l’accuracy ou le F1-score.

> ✅ Une **loss basse** ne garantit pas toujours une **métrique parfaite** (et inversement).


# III. 🧮🧮🧮 Fonctions de coût (Loss Function)

Après démonstration

## 🔵 Fonctions de coût utilisées dans AdaBoost

**AdaBoost** est une méthode de **classification binaire** qui repose sur l’agrégation de **modèles faibles pondérés**, en donnant davantage de poids aux erreurs commises.

--

### 1️⃣ Exponential Loss

La fonction de perte exponentielle est au cœur de l'algorithme AdaBoost :

$$
L(y, f(x)) = \exp(-y \cdot f(x))
$$

- $y \in \{-1, +1\}$ : vraie classe (codée en -1 ou +1)  
- $f(x)$ : somme pondérée des prédictions des modèles faibles  

🎯 **Pourquoi cette fonction ?**

- Elle donne **plus de poids aux erreurs** :  
  Lorsque $y \cdot f(x)$ est négatif (prédiction erronée), la fonction exponentielle prend des valeurs élevées.  
- Elle est **dérivable**, donc **optimisable** via des méthodes de gradient.  

> ✅ Fonction centrale du boosting d’origine (AdaBoost de Freund & Schapire, 1997)  
> ✅ Permet de corriger efficacement les erreurs de prédiction.  

---

## 🔵 Fonctions de coût utilisées dans Gradient Boosting

Gradient Boosting est une méthode flexible, applicable à des tâches de **régression** ou de **classification**, en fonction de la **fonction de coût** choisie.

---#

## 1️⃣ Régression – MSE (Mean Squared Error)

$$
L(y, \hat{y}) = (y - \hat{y})^2
$$

- 🎯 Objectif : pénaliser fortement les **grandes erreurs**
- $y$ = vraie valeur  
- $\hat{y}$ = valeur prédite

> Fonction classique en régression. Sensible aux ou#tliers.

---

## 2️⃣ Régression – MAE (Mean Absolute Error)

$$
L(y, \hat{y}) = |y - \hat{y}|
$$

- 🎯 Objectif : minimiser **l’erreur absolue**
- Moins sensible aux valeurs extrêmes (outliers)

> Utilisée quand on veut être **robuste aux valeurs atypiques**, mais moi#ns lisse que le MSE.

---

## 3️⃣ Classification – Log-loss 

$$
L(y, \hat{p}) = -y \log(\hat{p}) - (1 - y) \log(1 - \hat{p})
$$

- 🎯 Objectif : maximiser la probabilité associée à la **bonne classe**
- $y \in \{0, 1\}$ : vraie étiquette binaire  
- $\hat{p}$ = probabilité prédite d'appartenir à la classe 1

> Fonction utilisée pour la **classification binaire** (Gradient Boosting Classifier).

---

✅ En pratique, la fonction de coût choisie **définit la nature du problème** et influe directement sur **l’optimisation à chaque étape** du boosting.


# 🧠 Résumé des fonctions de coût

| Algorithme            | Loss principale utilisée           | Cas d'usage               |
|-----------------------|------------------------------------|---------------------------|
| **AdaBoost**          | Exponential loss                    | Classification binaire    |
| **Gradient Boosting** | MSE (Mean Squared Error)             | Régression                 |
|                       | Log-loss                            | Classification binaire        |
|                       | MAE (Least Absolute Deviation) (rare)| Régrutliers |

> 🎯 La fonction de coût sert à **apprendre**, la métrique sert à **évluer**.

````python
from sklearn.ensemble import AdaBoostClassifier
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report

# 📥 Charger un jeu de données de classification binaire
X, y = load_breast_cancer(return_X_y=True)

# ✂️ Séparer en train / test
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=42)

# ✅ Modèle AdaBoost (boosting par pondération des erreurs)
model = AdaBoostClassifier(
    n_estimators=50,        # 🌳 Nombre de modèles faibles (souvent des arbres de profondeur 1)
    learning_rate=1.0,      # ⚖️ Influence de chaque modèle faible (réduit pour régulariser)
    random_state=42         # 🧪 Pour reproductibilité
)

# 🎯 Apprentissage du modèle
model.fit(X_train, y_train)

# 📊 Prédiction et évaluation
y_pred = model.predict(X_test)
print("🔍 Rapport de classification (AdaBoost) :")
print(classification_report(y_test, y_pred))
````

$$

- Ce jeu de données est utile pour tester des modèles de régression capables de capturer des relations complexes, comme les **SVR avec noyaux non linéaires** (RBF, polynomial, etc.).


````python
from sklearn.ensemble import GradientBoostingRegressor, AdaBoostRegressor
from xgboost import XGBRegressor
from sklearn.datasets import make_friedman1
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error

# 📥 Générer un jeu de données de régression non linéaire (synthétique)
X, y = make_friedman1(n_samples=200, random_state=42)

# ✂️ Séparation en train / test
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=42)

# ✅ Modèle Gradient Boosting
model_gb = GradientBoostingRegressor(
    n_estimators=100,    # Nombre total d'arbres dans la séquence de boosting
    learning_rate=0.1,   # Facteur d'importance des nouvelles estimations (shrinkage)
    max_depth=3,         # Profondeur maximale des arbres (contrôle le surapprentissage)
    random_state=42      # Graine pour la reproductibilité
)

# ✅ Modèle XGBoost
model_xgb = XGBRegressor(
    n_estimators=100,     # Nombre total d'arbres à construire
    learning_rate=0.1,    # Taux d'apprentissage pour chaque arbre ajouté
    max_depth=3,          # Profondeur maximale des arbres (contrôle la complexité)
    random_state=42       # Graine pour la reproductibilité
)


# ✅ Modèle AdaBoost
model_ada = AdaBoostRegressor(
    n_estimators=100,    # Nombre total d'arbres (ou modèles de base)
    learning_rate=0.1,   # Facteur d'importance des erreurs corrigées
    random_state=42      # Graine pour la reproductibilité
)


# 🎯 Entraînement des modèles
model_gb.fit(X_train, y_train)
model_xgb.fit(X_train, y_train)
model_ada.fit(X_train, y_train)

# 📈 Prédiction
y_pred_gb = model_gb.predict(X_test)
y_pred_xgb = model_xgb.predict(X_test)
y_pred_ada = model_ada.predict(X_test)

# 📊 Évaluation
mse_gb = mean_squared_error(y_test, y_pred_gb)
mse_xgb = mean_squared_error(y_test, y_pred_xgb)
mse_ada = mean_squared_error(y_test, y_pred_ada)

print(f"📉 MSE (Gradient Boosting) = {mse_gb:.2f}")
print(f"📉 MSE (XGBoost) = {mse_xgb:.2f}")
print(f"📉 MSE (AdaBoost) = {mse_ada:.2f}")
````

## Introduire Validation croisée

### 🚦 Validation Croisée : Pourquoi ?

La validation croisée (cross-validation) permet de :

1. **Réduire la variance des estimations de performance :**
   - En évaluant le modèle sur plusieurs sous-échantillons, on obtient une mesure de performance plus fiable et moins dépendante d'un découpage particulier des données.

2. **Utiliser au mieux les données disponibles :**
   - Chaque observation est utilisée à la fois pour l'entraînement et pour le test (mais jamais en même temps), ce qui maximise l'exploitation du jeu de données.

3. **Détecter le surapprentissage :**
   - En testant sur des jeux de données différents, on évalue la capacité du modèle à généraliser, et non seulement à mémoriser les données d'entraînement.


````python
from sklearn.ensemble import GradientBoostingRegressor, AdaBoostRegressor
from xgboost import XGBRegressor
from sklearn.datasets import make_friedman1
from sklearn.model_selection import cross_val_score, KFold
from sklearn.metrics import make_scorer, mean_squared_error
import numpy as np

# 📥 Générer un jeu de données de régression non linéaire (synthétique)
X, y = make_friedman1(n_samples=200, random_state=42)

# ⚙️ Configuration de la validation croisée
kf = KFold(n_splits=5, shuffle=True, random_state=42)  # 5-fold cross-validation
scorer = make_scorer(mean_squared_error, greater_is_better=False)

# ✅ Modèle Gradient Boosting
model_gb = GradientBoostingRegressor(
    n_estimators=100, learning_rate=0.1, max_depth=3, random_state=42
)

# ✅ Modèle XGBoost
model_xgb = XGBRegressor(
    n_estimators=100, learning_rate=0.1, max_depth=3, random_state=42
)

# ✅ Modèle AdaBoost
model_ada = AdaBoostRegressor(
    n_estimators=100, learning_rate=0.1, random_state=42
)

# 🔄 Validation croisée
cv_scores_gb = cross_val_score(model_gb, X, y, cv=kf, scoring=scorer)
cv_scores_xgb = cross_val_score(model_xgb, X, y, cv=kf, scoring=scorer)
cv_scores_ada = cross_val_score(model_ada, X, y, cv=kf, scoring=scorer)

# 📊 Affichage des résultats
print(f"📉 MSE (Gradient Boosting) : {-np.mean(cv_scores_gb):.2f} ± {np.std(cv_scores_gb):.2f}")
print(f"📉 MSE (XGBoost) : {-np.mean(cv_scores_xgb):.2f} ± {np.std(cv_scores_xgb):.2f}")
print(f"📉 MSE (AdaBoost) : {-np.mean(cv_scores_ada):.2f} ± {np.std(cv_scores_ada):.2f}")
````

## 🌳 5.4 Prédiction finale par vote pondéré dans AdaBoost  

### 📝 Contexte :  
La prédiction finale dans AdaBoost est basée sur un **vote pondéré** des prédictions de chaque modèle faible.  
- Les modèles sont **pondérés en fonction de leur précision**.  
- Plus le poids est élevé, plus le modèle est **fiable**.  

---

### 🔧 Exemple concret : Prédiction d'un fruit inconnu  

| Poids (g) | Couleur (score 0–1) | Fruit prédit (Modèle 1) | Fruit prédit (Modèle 2) |
|----------|---------------------|-------------------------|-------------------------|
| 165      | 0.7                 | Pomme                   | Poire                    |  

#### 💡 Contexte des modèles :  
- **Modèle 1 (Couleur)** avec un poids de 0.3466  
- **Modèle 2 (Poids)** avec un poids de 0.568  
- Les poids reflètent la **fiabilité** des modèles :  
  - Plus le poids est élevé, plus le modèle est **influant** dans la décision finale.  

---

### 🔢 Étape 1 : Calcul des scores pondérés  
- Attribuer un **score positif** pour Pomme (+) et un **score négatif** pour Poire (-).  
- Prédiction du **Modèle 1** :  
  - Pomme → +0.3466  
- Prédiction du **Modèle 2** :  
  - Poire → -0.568  

#### Calcul du score global :  
$$
\text{Score final} = +0.3466 - 0.568 = -0.2214
$$  

---

### ✅ Interprétation du score :  
- Si le **score final est positif**, on prédit **Pomme**.  
- Si le **score final est négatif**, on prédit **Poire**.  
- Ici, le score est **-0.2214**, donc la prédiction finale est **Poire**.  

---

### 🌟 Pourquoi ce choix ?  
1. **Vote pondéré :**  
   - La **prédiction du modèle 2** a un poids plus élevé (0.568) que celle du **modèle 1** (0.3466).  
   - Le modèle 2 a donc **plus d'impact** dans la décision finale.  

2. **Robustesse d'AdaBoost :**  
   - En combinant les modèles avec des **poids différents**, AdaBoost prend en compte la **fiabilité relative** de chaque prédiction.  
   - Même si un modèle faible prédit mal, un modèle plus fiable peut **corriger l'erreur**.  

---

### 💡 En résumé :  
- La prédiction finale est obtenue en **combinant les prédictions pondérées** des modèles.  
- Si la **somme des scores est positive**, on prédit **Pomme**. Sinon, on prédit **Poire**.  
- AdaBoost utilise cette approche pour **renforcer l'impact des modèles les plus fiables**.
