# Arbres de décision (CART)

## 🌳 Pourquoi CART ?

**CART** signifie *Classification And Regression Trees*.  
C’est l’acronyme donné à l’algorithme d’arbres de décision proposé en 1986 par **Breiman, Friedman, Olshen et Stone**  
(Livre : *Classification and Regression Trees*).

---

### 🎯 Que permet CART ?

- Des tâches de **classification** (prédire une *catégorie*)  
- Des tâches de **régression** (prédire une *valeur numérique*)  

---

### 🧠 Idée générale de l’algorithme CART :

1. À chaque étape, **on teste toutes les coupures possibles** sur chaque variable.  
2. Chaque variable est testée **indépendamment**  
   → par exemple : 5 seuils × 3 variables = **15 coupures** à évaluer (et non 5 × 5 × 5).  
3. On coupe le nœud en **2 groupes** selon la coupure optimale.  
4. On choisit la coupure qui **maximise la pureté** (en classification)  
   ou **minimise l’erreur** (en régression).
   > Si les indices de Gini sont identiques, l'algorithme va souvent choisir la coupure la plus proche de la moyenne ou du centre de l'intervalle.
6. Le processus est **récursif** : il se répète sur chaque sous-groupe.  
7. L’arrêt du développement de l’arbre peut se produire si l’un des critères suivants est atteint :  
   - 🔹 Profondeur maximale atteinte  
   - 🔹 Nombre minimum d'observations dans un nœud  
   - 🔹 Gain de pureté trop faible  

---

✅ **En résumé** : CART construit un arbre en **découpant l’espace de façon adaptative**,  
en cherchant à **optimiser un critère simple à chaque étape**.  


## 1. Cas de la classification dans CART
## 🍏🍐 Exemple : classification Pommes et Poires

On dispose d’un petit jeu de données :

| Poids (g) | Diamètre (cm) | Couleur (score 0–1) | Fruit  |
|-----------|----------------|----------------------|--------|
| 150       | 7.5            | 0.7                  | Pomme  |
| 155       | 7.6            | 0.6                  | Poire  |
| 160       | 7.8            | 0.4                  | Poire  |
| 165       | 8.0            | 0.5                  | Pomme  |
| 170       | 8.1            | 0.8                  | Pomme  |
| 175       | 8.3            | 0.3                  | Poire  |

---

### 🌳 **Calcul des coupures possibles pour l'algorithme CART**  

Pour déterminer le nombre de coupures possibles avec l'algorithme **CART (Classification and Regression Trees)**, il faut examiner chaque variable numérique et identifier les **seuils de coupure** entre les valeurs distinctes.  

---

#### 📝 **Étapes pour calculer les coupures :**  

1. **Variables candidates :**  
   - **Poids (g) :** 6 valeurs distinctes (150, 155, 160, 165, 170, 175)  
   - **Diamètre (cm) :** 6 valeurs distinctes (7.5, 7.6, 7.8, 8.0, 8.1, 8.3)  
   - **Couleur (score 0–1) :** 6 valeurs distinctes (0.3, 0.4, 0.5, 0.6, 0.7, 0.8)  

Pour chaque variable avec $n$ valeurs distinctes, il y a **$n-1$ coupures possibles**.  
- **Poids :** $6-1=5$ coupures (Seuils : 152.5, 157.5, 162.5, 167.5, 172.5)  
- **Diamètre :** $6-1=5$ coupures (Seuils : 7.55, 7.7, 7.9, 8.05, 8.2)  
- **Couleur :** $6-1=5$ coupures (Seuils : 0.35, 0.45, 0.55, 0.65, 0.75)  


### 🌳 **Première itération : à la racine**  
Au début de la première itération, l'algorithme teste **toutes les coupures possibles sur chaque variable** pour trouver celle qui **minimise l'indice de Gini** ou maximise le **gain d'information**.  
Il choisit ensuite la **coupure optimale** et divise les données en deux sous-ensembles pour former les branches de l'arbre.  

#### 🪓 **Choix de la coupure optimale :**  
On teste toutes les coupures possibles sur chaque variable pour minimiser l’indice de Gini.  

**Exemple :**  
Si la meilleure coupure est **Couleur > 0.65**, alors :  
- **À droite (Couleur > 0.65)** → 0.7 (Pomme), 0.8 (Pomme) → Pur (Pommes)  
- **À gauche (Couleur ≤ 0.65)** → 0.3 (Pomme), 0.4 (Pomme), 0.5 (Pommes), 0.6 (Poire)  

#### 🌳 Calcul de l'indice de Gini pour une coupure  

Quand on coupe un nœud en deux sous-groupes (gauche et droite), le Gini total après la coupure est donné par :  

$$
Gini_{split} = \frac{n_{\text{gauche}}}{n} \cdot Gini(\text{gauche}) + \frac{n_{\text{droite}}}{n} \cdot Gini(\text{droite})
$$

Où :  
- $n_{\text{gauche}}$ = nombre d'exemples dans le sous-groupe gauche  
- $n_{\text{droite}}$ = nombre d'exemples dans le sous-groupe droit  
- $n$ = nombre total d'exemples dans le nœud initial  

---

### 🔢 Formule de l'indice de Gini pour chaque groupe  

Pour calculer l'**indice de Gini** d'un groupe, on utilise la formule suivante :  

$$
Gini = 1 - \sum_{i=1}^{C} p_i^2
$$

Où :  
- $C$ = nombre de classes (ici, 1 pour Pomme)  
- $p_i$ = proportion d'éléments de la classe $i$ dans le groupe  

---

### ✅ **Interprétation :**  
- Un indice de Gini de **0** indique un groupe **parfaitement pur**, c’est-à-dire que tous les exemples appartiennent à la même classe.  
- Plus l'indice de Gini est élevé, plus le groupe est **hétérogène**.  
- L'objectif de l'algorithme CART est de **minimiser l'indice de Gini** pour obtenir des sous-groupes aussi homogènes que possible.  
- Lors de chaque coupure, l'algorithme choisit celle qui réduit le plus l'indice de Gini, ce qui conduit à une séparation plus nette entre les classes.  


### 🌳 Coupure : `Couleur > 0.65`  
Cette coupure sépare les données en deux groupes :  

- **Groupe droit** (Couleur > 0.65) → 0.7 (Pomme), 0.8 (Pomme)
  - 2 exemples : 2 Pommes  
- **Groupe gauche** (Couleur ≤ 0.65) → 0.3 (Pomme), 0.4 (Pomme), 0.5 (Pommes), 0.6 (Poire)  
  - 4 exemples : 1 Pomme, 3 Poires


#### 🔢 Probabilités dans le groupe gauche :  
Le calcul des probabilités pour chaque classe (Pomme et Poire) dans le **groupe gauche** est :  
$$
p(\text{Pomme}) = \frac{1}{4}, \quad p(\text{Poire}) = \frac{3}{4}
$$


### 🌳 Calcul pour le groupe droit (Couleur > 0.65)  

Le groupe droit contient **2 exemples** :  
- **2 Pomme**  
- **0 Poire**  

#### 📝 **Étape 1 : Calcul des proportions :**  

$$
p(\text{Pomme}) = \frac{2}{2} = 1, \quad p(\text{Poire}) = \frac{0}{2} = 0
$$

#### 🔢 **Étape 2 : Appliquer la formule de Gini :**  

$$
Gini = 1 - \left[p(\text{Pomme})^2 + p(\text{Poire})^2\right]
$$  

$$
Gini = 1 - [1^2 + 0^2] = 1 - (1 + 0) = 0
$$  

---

### ✅ **Interprétation :**  
Un **indice de Gini de 0** indique que le groupe est **parfaitement pur**, c’est-à-dire qu’il contient des exemples d’une seule classe (ici, uniquement des Pommes).  
Cela signifie que ce groupe est **idéalement séparé** du point de vue de la classification.  



### 🌳 Calcul du Gini à gauche :

$$
\begin{aligned}
Gini(\text{gauche}) &= 1 - \left(\frac{1}{4}\right)^2 - \left(\frac{3}{4}\right)^2 \\
&= 1 - \frac{1}{16} - \frac{9}{16} \\
&= 1 - \frac{10}{16} \\
&= \frac{6}{16} = 0.375
\end{aligned}
$$

### 🔢 Calcul final  

$$
Gini_{split} = \frac{4}{6} \cdot 0.375 + \frac{2}{6} \cdot 0 = 0.25
$$


---

## 🔍 Comparaison des autres coupures (résumé)

| Variable   | Seuil testé     | Gini obtenu |
|------------|------------------|-------------|
| Couleur    | 0.35             | 0.444       |
| Couleur    | 0.45             | 0.444       |
| Couleur    | 0.55             | 0.278       |
| Couleur    | 0.65             | **0.250** ✅ |
| Couleur    | 0.75             | 0.444       |

---

## 🧠 Conclusion

La coupure `Couleur > 0.65` est la meilleure ici (Gini = 0.250 ;  Gini initial = 0.5).  
Elle produit un groupe **pur à droite** (100 % Pommes), et un groupe encore mixte à gauche.

On poursuivra ensuite l'algorithme **en coupant le groupe gauche** dans la prochaine itération.


## 🔁 Deuxième itération (sous-arbre gauche car droite est déjà pur)

On continue avec le sous-ensemble : arbre gauche

#### ➤ Sur Poids (valeurs : 155, 160, 165, 175)  
➔ seuils = 157.5, 162.5, 170

#### ➤ Sur Diamètre (valeurs : 7.6, 7.8, 8.0, 8.3)  
➔ seuils = 7.7, 7.9, 8.15

#### ➤ Sur Couleur (valeurs : 0.3, 0.4, 0.5, 0.6)  
➔ seuils = 0.35, 0.45, 0.55

✅ Encore 9 coupures testées !

---

## 🔚 On continue ainsi jusqu’à obtenir :
- des groupes **suffisamment purs**
- ou un **critère d’arrêt** atteint (ex : profondeur max, nombre min d’échantillons…)
- cela dépend des **hyperparamètres du modèle**


| **Hyperparamètre**         | **Rôle**                                                        | **Exemple**                      |
|----------------------------|------------------------------------------------------------------|----------------------------------|
| `max_depth`                | Profondeur maximale de l’arbre                                  | `max_depth = 5`                 |
| `min_samples_split`        | Nombre min d’observations pour tenter une coupure               | `min_samples_split = 2`         |
| `min_samples_leaf`         | Nombre min d’observations dans une feuille terminale            | `min_samples_leaf = 1`          |
| `min_impurity_decrease`    | Gain minimal de pureté pour accepter une coupure                | `min_impurity_decrease = 0.0001`|

### 🌳 **Coupures possibles pour 4 catégories : Exemple avec CART**  

Pour une variable catégorielle avec **4 couleurs** (Red, Yellow, Blue, Green), les coupures se font en divisant l'ensemble en **deux groupes non vides et distincts**.  

#### 💡 **Nombre de combinaisons binaires :**  
- Pour une variable avec $k$ catégories, le nombre de **coupures possibles** est donné par :  
  $$
  \text{Nombre de coupures} = \frac{2^k - 2}{2}
  $$  
  - $2^k$ : nombre total de sous-ensembles (y compris vides)  
  - **-2** : pour exclure les ensembles vides à gauche ou à droite  
  - **÷2** : pour ne pas compter les **mêmes partitions inversées**  

#### 📝 **Application pour 4 couleurs :**  
- $2^4 - 2 = 14$ (nombre total de sous-ensembles non vides)  
- Diviser par 2 pour ne pas compter les mêmes partitions inversées :  
  $$
  \frac{14}{2} = 7 \text{ coupures possibles}
  $$  

---

### 🌈 **Les 7 coupures possibles :**  
1. **Red** | **Yellow, Blue, Green**  
2. **Yellow** | **Red, Blue, Green**  
3. **Blue** | **Red, Yellow, Green**  
4. **Green** | **Red, Yellow, Blue**  
5. **Red, Yellow** | **Blue, Green**  
6. **Red, Blue** | **Yellow, Green**  
7. **Red, Green** | **Yellow, Blue**  

---

### 📝 **Meilleure coupure trouvée par CART :**  
- **Coupure optimale** : **{Red, Yellow} | {Blue, Green}**  
- **Gini** : **0.0** (parfaitement pur)  

#### 🔍 **Pourquoi cette coupure ?**  
- Cette coupure sépare **les pommes (Red, Yellow)** des **poires (Blue, Green)**, ce qui **maximise la pureté** (Gini = 0).  
- Les deux groupes sont **parfaitement homogènes**, avec uniquement des pommes d'un côté et des poires de l'autre.  

---

### ✅ **Conclusion :**  
La coupure optimale pour séparer les **fruits par couleur** est **{Red, Yellow} | {Blue, Green}**,  
car elle permet d'obtenir des sous-groupes **parfaitement purs** avec un indice de Gini égal à **0**.  


## 2. Cas de la régression dans CART

Lorsque la tâche est de **prédire une valeur continue** (régression), CART cherche à **minimiser la variance** dans chaque sous-groupe créé par une coupure.

Formule de la variance dans un nœud :

\\[
Var(t) = \frac{1}{n_t} \sum_{i=1}^{n_t} (y_i - \bar{y}_t)^2
\\]

## Exemple : prédire la note d'un étudiant

| Temps de révision (heures) | Note sur 20 |
|:---------------------------:|:-----------:|
| 2                           | 10          |
| 3                           | 12          |
| 4                           | 14          |
| 5                           | 15          |
| 7                           | 18          |
| 8                           | 19          |

## 🔢 Et en régression ? Comment l’arbre prédit une valeur Y ?

Un **arbre de décision en régression** fonctionne un peu différemment de la classification :

---

### 🧠 Principe :

1. L’arbre découpe l’espace des variables $X$ en sous-groupes (feuilles).
2. Chaque feuille contient les exemples d’entraînement qui tombent dans ce sous-espace.
3. Lorsqu’un nouvel exemple arrive, il est envoyé dans une feuille.
4. ✅ **La prédiction est la moyenne des $Y$ d’entraînement dans cette feuille.**

---

### ✏️ Exemple :

Imaginons que dans une feuille finale de l’arbre, on a ces données :

- `temps` entre 5h et 7h
- `notes` observées = [15, 18]

Alors, pour un nouvel étudiant avec `temps = 6h`, la prédiction de l’arbre sera :

$$
\hat{y} = \frac{15 + 18}{2} = 16.5
$$

---

> 🔍 Il n’y a **pas de formule** ou d’équation complexe.  
> L’arbre **moyenne simplement les cibles** des données dans la feuille correspondante.

---

✅ C’est pour cela qu’on dit qu’un **arbre de décision pour régression** produit une **fonction en escalier** (valeurs constantes par zones).


## ✂️ Étapes :

1. **Coupure proposée** : `temps < 5` heures.

---

2. **Groupes formés** :

- **Groupe 1** : notes 10, 12, 14, 15  
  → moyenne = 12.75  
  → variance :

$$
\text{Variance Groupe 1} =
\frac{(10 - 12.75)^2 + (12 - 12.75)^2 + (14 - 12.75)^2 + (15 - 12.75)^2}{4}
= 3.6875
$$

---

- **Groupe 2** : notes 18, 19  
  → moyenne = 18.5  
  → variance :

$$
\text{Variance Groupe 2} = \frac{(18 - 18.5)^2 + (19 - 18.5)^2}{2} = 0.25
$$

---

3. **Calcul de la variance pondérée après coupure** :

$$
\text{Var}_{\text{après}} = \frac{4}{6} \times 3.6875 + \frac{2}{6} \times 0.25 = 2.5417
$$

---

4. **Moyenne globale** :

$$
\text{Moyenne} = \frac{10 + 12 + 14 + 15 + 18 + 19}{6} = 14.6667
$$

5. **Variance totale avant coupure** :

$$
\text{Var}_{\text{avant}} =
\frac{(10 - 14.67)^2 + (12 - 14.67)^2 + (14 - 14.67)^2 + (15 - 14.67)^2 + (18 - 14.67)^2 + (19 - 14.67)^2}{6}
\approx 11.22
$$

---

6. **Gain de variance** :

$$
\text{Gain} = 11.22 - 2.5417 = 8.6783
$$

✅ **Cette coupure réduit beaucoup la variance ➔ elle sera choisie.**


## 🤔 Est-ce qu’un petit gain suffit pour couper ? (ex : gain = 1)

La décision de garder une coupure **ne dépend pas uniquement** de la taille du gain de variance.  
Elle dépend aussi des **hyperparamètres fixés** pour contrôler le développement de l’arbre.

---

### 🎯 Rappel : que fait CART à chaque étape ?

- Il calcule la **variance avant** coupure.
- Il teste toutes les coupures possibles.
- Il calcule la **variance après** chaque coupure.
- Il retient celle qui donne le **plus grand gain de variance** :
  
  $$
  \text{Gain} = \text{Var}_{\text{avant}} - \text{Var}_{\text{après}}
  $$

---

### 📌 Est-ce que le gain est “suffisant” ?

| Gain obtenu | Coupure retenue ? | Dépend de…                           |
|-------------|--------------------|--------------------------------------|
| Grand gain  | ✅ Oui              | Très probable                        |
| Petit gain  | ⚠️ Peut-être        | Si supérieur à `min_impurity_decrease` |
| Gain nul    | ❌ Non              | Pas de réduction, donc rejeté        |

---

### ⚙️ Le rôle du paramètre `min_impurity_decrease`

Dans `sklearn`, ce paramètre (appelé ainsi même en régression) fixe un **seuil minimum** pour considérer qu’une coupure “vaut le coup”.

> 🔹 Par défaut, ce seuil est `0`, donc **toute amélioration est acceptée**.  
> 🔹 Si on fixe `min_impurity_decrease = 1.5`, alors un **gain de 1.0 sera ignoré**, même s’il est mathématiquement une amélioration.

---

### ✅ Conclusion

Un **gain élevé** est généralement un bon signe.  
Mais pour qu’une coupure soit réellement **retenue**, le gain doit être **supérieur au seuil fixé par l’utilisateur** (ou égal à zéro s’il n’y a aucun seuil).

Cela permet à l’algorithme d’éviter de couper pour de **toutes petites améliorations** sans réel intérêt.


## ⚠️ Pourquoi risque-t-on du surapprentissage (overfitting) ?

- **Si l’arbre pousse trop profondément**, il finit par **mémoriser le bruit** dans les données.
- Il **s’adapte parfaitement aux anomalies**, aux erreurs ponctuelles ou cas rares.

---

### 🎯 Conséquences :

- Très **bonne performance sur les données d'entraînement**  (*train*)=> ne sert à rien
- Mais **mauvaise capacité à généraliser** sur de nouvelles données (*test*)

---

Un arbre sans limitation a :

- Un **biais très faible** (il colle parfaitement aux données)
- Mais une **variance très élevée** (très sensible aux fluctuations)


## 💡 Exemple illustré

Supposons le jeu de données suivant :

| Temps de révision (heures) | Note sur 20 |
|----------------------------|-------------|
| 2                          | 10          |
| 3                          | 12          |
| 4                          | 14          |
| 5                          | 17          |
| 6                          | **1** ⚠️ *(anomalie)* |
| 7                          | 18          |
| 8                          | 19          |

---

- La relation générale semble être : **plus on révise, meilleure est la note**.
- Mais il y a ici **une anomalie à 6h** (score très bas de **1**).

---

### 🌲 Si l’arbre pousse librement :

- Il pourrait créer une **coupure spéciale juste pour s’adapter à l’anomalie** à **6h**, traitant cet exemple comme une exception.
- Cela entraînerait un **arbre trop complexe** qui **sur-adapte aux données d'entraînement**, capturant cette anomalie plutôt que la tendance générale.

---

### ✅ **Or, cette anomalie n’est pas représentative du phénomène réel.**  
L'arbre doit **mieux généraliser** et ne pas ajuster ses prédictions uniquement sur des cas extrêmes ou des **valeurs aberrantes**.

> 🔍 **Solution :** Les arbres de régression doivent être régularisés ou limités pour éviter ce genre de sur-apprentissage (par exemple, en limitant la profondeur de l'arbre ou en utilisant un nombre minimal d'échantillons pour diviser un nœud).


## Résumé

| Concept | Résultat |
|---------|----------|
| Arbre pousse sans limitation | Apprend aussi le bruit |
| Trop bien adapté aux données d'entraînement | Overfitting |
| Mauvaise capacité à généraliser | Mauvaise performance sur test |

**Moralité** : contrôler la taille et la profondeur des arbres est indispensable pour éviter l'overfitting.
