Fléau de la dimension : illustration

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
mat = pd.read_csv('mat.txt', header=None, index_col=0, sep=';')
```

Je définis une grille

```python
grille = list(range(2, 10)) + list(range(10, 91, 10)) + list(range(100, 1001, 100))
res = []
```

Je calcule le rapport max/min des normes

```python
for g in grille:
    tmp = np.linalg.norm(mat.iloc[:, :g], axis=1)
    res.append(max(tmp) / min(tmp))
```

Je dessine

```python
plt.figure()
plt.plot(grille, res, 'b-', marker='o', color='red')
plt.title('Evolution du contraste')
plt.xlabel('Grille')
plt.ylabel('Rapport max/min des normes')
plt.show()
```
