## 1. 🧠 Neurone formel : principe de base

Un **neurone formel** simule le comportement d’un neurone biologique :  
il prend plusieurs **entrées pondérées**, les additionne, puis applique une **fonction d’activation** pour produire une sortie.

#### 🔧 Formules :

$$
z = w_1 x_1 + w_2 x_2 + \dots + w_n x_n + b
$$

$$
\hat{y} = f(z)
$$

- \( w_i \) : poids associé à l’entrée \( x_i \)  
- \( b \) : biais (comporte comme un décalage)  
- \( f \) : fonction d’activati on appliquée à la somme pondérée  


![Description de l'image](Architecture_MPL.webp)

- Si vous avez, par exemple, un jeu de données avec trois caractéristiques (features) $x_1$, $x_2$, et $x_3$, alors chaque cercle de la couche d'entrée correspondra à l'une de ces variables. Ces entrées sont ensuite transmises aux neurones de la couche cachée via des connexions pondérées.

- Chaque neurone est bien une combinaison linéaire (une régression) suivie d'une non-linéarité (fonction d'activation).

- Vous pouvez avoir autant de couches cachées que vous le souhaitez dans un réseau de neurones. 

## 🧩 Chaque neurone apprend ses propres poids

- Tous les neurones d’une même couche reçoivent *les mêmes entrées* $x$,  
  mais *chacun apprend ses propres poids* $w$ et biais $b$.  
- Cela permet à chaque neurone de *capturer une caractéristique différente* des données.  

#### 💡 Exemple pour deux neurones dans la même couche :

- Neurone 1 :  
$$
z_1 = w_{11}x_1 + w_{12}x_2 + \dots + b_1
$$

- Neurone 2 :  
$$
z_2 = w_{21}x_1 + w_{22}x_2 + \dots + b_2
$$

➡️ Ainsi, *chaque neurone est indépendant dans son apprentissage*, bien qu’il travaille sur les mêmes données d’entrée.  

---

### 💡 Neurone et régression linéaire

- Un neurone formel réalise une *combinaison linéaire* des entrées, semblable à une *régression linéaire*.  
- La sortie est calculée comme :  
$$
z = \sum (w_i \cdot x_i) + b
$$  
- Si la fonction d'activation est une *fonction identité* f(x) = x, le neurone devient exactement une *régression linéaire*.  
- Cependant, en utilisant une *fonction d'activation non linéaire* (comme *ReLU* ou *sigmoïde*), le neurone est capable de *modéliser des relations complexes*.  

➡️ Chaque neurone peut être vu comme une *petite régression linéaire*, mais l'ensemble du réseau devient un *modèle complexe* grâce aux couches cachées et aux activations.  


## 2. 🔹 Perceptron Multicouche (MLP)

Un MLP est un réseau composé de :
- une couche d'entrée,
- une ou plusieurs couches cachées,
- une couche de sortie.

Chaque couche est composée de plusieurs *neurones*, chacun appliquant une *combinaison linéaire* des entrées suivie d'une *fonction d'activation*.

➡️ Le MLP utilise cette structure pour apprendre des *représentations complexes* des données en passant par plusieurs couches intermédiaires (cachées).  
- Les *couches cachées* permettent de *capturer des motifs* de plus en plus complexes.  
- La *couche de sortie* donne la *prédiction finale* en appliquant une dernière activation.  

💡 Contrairement aux réseaux simples (comme le perceptron monocouche), le MLP peut modéliser des *relations non linéaires* grâce à ses *couches cachées et fonctions d'ac
tivation*.  


## 🌟 3. Fonctions d'activation classiques

### 1. Fonctions d'activation pour les couches cachées

#### **ReLU (Rectified Linear Unit)**
$$
\text{ReLU}(z) = \max(0, z)
$$
- Utilisée pour éviter le gradient évanescent.  
- **Sortie** : entre $0$ et $+\infty$.  
- **Application :** Réseaux profonds (CNN, DNN).  

---

#### **Leaky ReLU**
$$
\text{Leaky ReLU}(z) =
\begin{cases} 
z & \text{si } z > 0 \\
\alpha z & \text{si } z \leq 0
\end{cases}
$$
- Variante de ReLU pour éviter les neurones morts.  
- **Sortie** : $(-\infty, +\infty)$.  
- **Application :** Réseaux profonds avec risque de neurones morts.  

---

#### **ELU (Exponential Linear Unit)**
$$
\text{ELU}(z) =
\begin{cases} 
z & \text{si } z > 0 \\
\alpha (e^{z} - 1) & \text{si } z \leq 0
\end{cases}
$$
- Fonction lisse et asymétrique, utile pour de meilleures performances.  
- **Sortie** : $(-\alpha, +\infty)$.  
- **Application :** Réseaux profonds où la stabilité est importante.  

---

#### **Tangente hyperbolique (tanh)**
$$
\tanh(z) = \frac{e^{z} - e^{-z}}{e^{z} + e^{-z}}
$$
- Centrée sur zéro, utile pour normaliser les données.  
- **Sortie** : entre $-1$ et $1$.  
- **Application :** Réseaux peu profonds ou lorsque les données sont normalisées.  

---

### 2. Fonctions d'activation pour la couche de sortie (n’est pas un hyperparamètre)

#### **Sigmoïde**
$$
\sigma(z) = \frac{1}{1 + e^{-z}}
$$
- Utilisée pour les problèmes de classification binaire.  
- **Sortie** : entre $0$ et $1$.  
- **Application :** Prédiction de probabilité pour une classe (binaire).  

---

#### **Softmax**
$$
\text{Softmax}(z_i) = \frac{e^{z_i}}{\sum_{j=1}^{n} e^{z_j}}
$$
- Transforme les sorties en probabilités pour chaque classe (multiclasse).  
- **Sortie** : somme des sorties = $1$.  
- **Application :** Classification multi-classes.  

---

#### **Activation linéaire (ou sans activation)**
$$
f(z) = z
$$
- Utilisée pour les problèmes de régression.  
- **Sortie** : valeur continue non bornée.  
- **Application :** Prédiction de valeurs numériques (prix, températures, etc.).  

---

### 💡 Remarques
- Les fonctions **ReLU**, **Leaky ReLU** et **ELU** sont principalement utilisées dans les **couches cachées** pour capturer des non-linéarités et éviter le gradient évanescent.  
- Les fonctions **Sigmoïde** et **Softmax** sont spécifiques aux **couches de sortie** pour produire des **probabilités**.  
- **Linéaire** est utilisée en sortie pour les **problèmes de régression**.  
- Les fonctions **sigmoïde** et **tanh** peuvent poser des problèmes de **gradient évanescent** dans les réseaux profonds.  
- Les variantes de ReLU (**Leaky ReLU**, **ELU**) sont souvent préférées pour leur robustesse.  


### 💡 Choix des fonctions d'activation selon le type de réseau

| Type de réseau         | Fonction d'activation recommandée       | Alternatives                                      |
|-----------------------|----------------------------------------|---------------------------------------------------|
| Réseau profond (CNN, DNN) | **ReLU**                              | Leaky ReLU, ELU (si neurones morts)                 |
| Réseau récurrent (LSTM, GRU) | **tanh** pour les états cachés       | Sigmoïde pour les portes (gates)                    |
| Réseau peu profond     | **tanh**                                | ReLU si les données sont non centrées               |
| Régression            | **Linéaire** (sortie)                    | Aucune activation                                  |
| Classification binaire| **Sigmoïde** (sortie)                    | -                                                 |
| Classification multi-classes | **Softmax** (sortie)              | -                                                 |


### 🧠 Fonction d'activation différente par couche : théorie et pratique

#### 💡 Théoriquement :
Dans les réseaux de neurones, il est **possible d'utiliser des fonctions d'activation différentes** pour chaque couche cachée.  
Cela peut être utile pour **capturer des caractéristiques complexes** en combinant les avantages de chaque fonction.

#### ⚠️ Implémentation dans Scikit-learn :
- Dans **Scikit-learn**, le modèle `MLPClassifier` utilise **une seule fonction d'activation** pour **toutes les couches cachées**.  
- Il n'est **pas possible de spécifier des fonctions d'activation différentes** pour chaque couche.  
- Cela simplifie l'implémentation et assure la **cohérence du réseau**, mais limite la personnalisation.  

---

### 🚀 Utilisation avec d'autres frameworks :
Pour des configurations plus complexes, il est préférable d'utiliser des frameworks comme **TensorFlow/Keras** ou **PyTorch**, qui permettent de spécifier une fonction d'activation **par couche**.  

##### Exemple avec Keras :
```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

model = Sequential([
    Dense(100, activation='relu', input_shape=(64,)),  # Première couche cachée avec ReLU
    Dense(50, activation='tanh'),                      # Deuxième couche cachée avec Tanh
    Dense(10, activation='softmax')                     # Couche de sortie pour la classification
])


# 🧠 Réseau de neurones : prédire si une personne va courir

---

## 🔹 1. Contexte

On souhaite prédire si une personne va **courir** ou non, en fonction de la **température** et de l’**humidité**.

- $x_1 = 10$ (Température en °C)
- $x_2 = 80$ (Humidité en %)
- Vérité terrain : $y = 1$ (la personne a couru)

Le réseau est composé de :

- 2 **neurones cachés** (fonction d'activation : sigmoïde),
- 1 **neurone de sortie** (sigmoïde),
- Apprentissage via rétropropagation.

---

## 🔹 2. Initialisation des poids

Les poids sont initialisés aléatoirement à l’aide d’une **distribution uniforme**.  
Par exemple, avec la méthode Xavier :

$$
w_i \sim \mathcal{U}\left( -\sqrt{\frac{6}{n_{\text{in}} + n_{\text{out}}}},\ \sqrt{\frac{6}{n_{\text{in}} + n_{\text{out}}}} \right)
$$

Si $n_{\text{in}} = 2$, $n_{\text{out}} = 2$ :

$$
w_i \sim \mathcal{U}(-1.22,\ +1.22)
$$

Dans notre exemple, on suppose que l'initialisation a donné les poids suivants :

### Couche cachée :

- $w_{11} = 0.2$, $w_{12} = 0.4$, $b_1 = 0.0$
- $w_{21} = -0.3$, $w_{22} = 0.1$, $b_2 = 0.0$

### Couche de sortie :

- $w_3 = 0.6$, $w_4 = -0.1$, $b_3 = 0.0$

Taux d’apprentissage : $\eta = 0.1$

---

## 🔹 3. Forward pass

### Formules générales

- Pour un neurone :  
  $$
  z = \sum w_i x_i + b
  $$  
  $$
  \hat{y} = \sigma(z) = \frac{1}{1 + e^{-z}}
  $$

---

### 🔸 Neurone caché 1 :

$$
z_1 = w_{11} \cdot x_1 + w_{12} \cdot x_2 + b_1 = 0.2 \cdot 10 + 0.4 \cdot 80 + 0 = 34
$$

$$
\hat{y}_1 = \sigma(34) \approx 1.0
$$

---

### 🔸 Neurone caché 2 :

$$
z_2 = w_{21} \cdot x_1 + w_{22} \cdot x_2 + b_2 = -0.3 \cdot 10 + 0.1 \cdot 80 + 0 = 5
$$

$$
\hat{y}_2 = \sigma(5) \approx 0.9933
$$

---

### 🔸 Neurone de sortie :

$$
z_3 = w_3 \cdot \hat{y}_1 + w_4 \cdot \hat{y}_2 + b_3
= 0.6 \cdot 1.0 - 0.1 \cdot 0.9933 + 0 = 0.5007
$$

$$
\hat{y} = \sigma(0.5007) \approx 0.6226
$$

---

## 🔹 4. Backward pass

### Formule de l’erreur en sortie :

$$
\delta_{\text{sortie}} = (\hat{y} - y) \cdot \hat{y} \cdot (1 - \hat{y})
$$

---

### Calcul :

$$
\delta_3 = (0.6226 - 1) \cdot 0.6226 \cdot (1 - 0.6226) \approx -0.0888
$$

---

### Formule de mise à jour des poids :

$$
\Delta w = -\eta \cdot \delta \cdot \text{entrée}
$$

---

### 🔸 Mise à jour des poids de sortie :

$$
\Delta w_3 = -0.1 \cdot (-0.0888) \cdot \hat{y}_1 = 0.00888
$$

$$
\Delta w_4 = -0.1 \cdot (-0.0888) \cdot \hat{y}_2 \approx 0.00882
$$

$$
\Delta b_3 = -0.1 \cdot (-0.0888) = 0.00888
$$

**Poids mis à jour** :

- $w_3 = 0.60888$
- $w_4 = -0.09118$
- $b_3 = 0.00888$

---

## 🔹 5. Propagation vers la couche cachée

### Formule :

$$
\delta_{\text{hidden}} = \delta_{\text{next}} \cdot w \cdot \hat{y}(1 - \hat{y})
$$

---

### 🔸 Neurone caché 2 :

$$
\delta_2 = \delta_3 \cdot w_4 \cdot \hat{y}_2 \cdot (1 - \hat{y}_2)
= -0.0888 \cdot (-0.1) \cdot 0.9933 \cdot 0.0067 \approx 0.000059
$$

$$
\Delta w_{21} = -0.1 \cdot 0.000059 \cdot 10 = -0.000059
$$

$$
\Delta w_{22} = -0.1 \cdot 0.000059 \cdot 80 = -0.000472
$$

---

### 🔸 Neurone caché 1 :

$$
\delta_1 = \delta_3 \cdot w_3 \cdot \hat{y}_1 \cdot (1 - \hat{y}_1)
= -0.0888 \cdot 0.6 \cdot 1.0 \cdot (1 - 1.0) = 0
$$

> ⚠️ Pas de mise à jour pour le neurone 1 car sa sortie est **saturée** à 1, donc $\sigma'(z) = 0$

---

## ✅ 6. Conclusion pédagogique

- Les poids sont **initialisés aléatoirement** dans un intervalle contrôlé.
- Les **activations** sont calculées à l’aide de la **sigmoïde**.
- La prédiction finale ($\hat{y} = 0.62$) est comparée au vrai label ($y = 1$).
- Une **rétropropagation complète** est faite :
  - la **couche de sortie** est bien mise à jour,
  - la **couche cachée** est partiellement mise à jour : seul le neurone 2 a pu bouger.


## 4. 💻 Exercice pratique : classification de chiffres manuscrits

On va utiliser un MLP pour reconnaître des chiffres à partir du jeu de données `digits` de `sklearn`.
### 4.1 MLP sans tuning


````python
from sklearn.datasets import load_digits
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import accuracy_score
````

````python
digits = load_digits()
digits

binary_mark = (digits.target ==2)|(digits.target ==3)
X, y = digits.data[binary_mark], digits.target[binary_mark]

y = (y == 3).astype(int)

X_train, X_test, y_train, y_test = train_test_split(X, y, random_state= 42)

mlp_basic = MLPClassifier(random_state = 33)
y_pred_basic = mlp_basic.fit(X_train, y_train).predict(X_test)

print(accuracy_score(y_test, y_pred_basic))

precision_score(y_test, y_pred_basic)

f1_score(y_test, y_pred_basic)
````

### 4.2 MLP avec Tuning

````python
from sklearn.model_selection import GridSearchCV

# Paramètres pour GridSearchCV
param_grid = {
    'hidden_layer_sizes': [(50,), (100,), (50, 50)],
    'activation': ['relu', 'tanh'],
    'solver': ['adam', 'sgd'], # l'algorithme d'optimisation utilisé pour mettre à jour les poids lors de l'entraînement du réseau de neurones
    'alpha': [0.0001, 0.001], # dans sklearn, régularisation est toujours ridge, alpha correspond à l'intensité de la pénalité
}

# Modèle MLP avec GridSearchCV
mlp_tuned = MLPClassifier(random_state=0)
grid_search = GridSearchCV(mlp_tuned, param_grid, cv=3, scoring='accuracy')
grid_search.fit(X_train, y_train)

# Meilleurs paramètres et score
print("\nMeilleurs paramètres :", grid_search.best_params_)
print("Meilleur score (train) :", grid_search.best_score_)

# Prédictions avec le meilleur modèle
y_pred_tuned = grid_search.best_estimator_.predict(X_test)
print("\nMLP avec tuning")
print("Accuracy :", accuracy_score(y_test, y_pred_tuned))
# print(classification_report(y_test, y_pred_tuned))
````
