# 🌟 Gradient Boosting – Regression en Deux Itérations

## 1. 📊 Données

| Étudiant | $x$ (heures de révision) | $y$ (note réelle) |
|----------|------------------------|------------------|
| A        | 2                      | 10               |
| B        | 4                      | 14               |
| C        | 6                      | 18               |

---

## 2. 🔁 Première itération

### 🔹 Étape 1 : Initialisation

On commence avec une prédiction constante : la moyenne des notes.

$$\hat{y}^{(0)} = \frac{10 + 14 + 18}{3} = 14$$  
> 🔎 La prédiction initiale est la moyenne des valeurs réelles.

---

### 🔹 Étape 2 : Calcul explicite de la fonction de perte

Dans cet exemple de boosting, nous utilisons la **fonction de perte MSE (Mean Squared Error)** pour mesurer l'erreur globale :
$$
\text{MSE} = \frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2
$$

#### 📝 Calcul de la MSE initiale :

| Étudiant | $y$ | $\hat{y}^{(0)}$ | $(y_i - \hat{y}^{(0)})^2$ |
|---------|-----|----------------|----------------------------|
| A       | 10  | 14             | $(-4)^2 = 16$               |
| B       | 14  | 14             | $(0)^2 = 0$                 |
| C       | 18  | 14             | $(+4)^2 = 16$               |

$$
\text{MSE} = \frac{1}{3} (16 + 0 + 16) = \frac{32}{3} \approx 10.67
$$

> 🔎 La **perte initiale** est relativement élevée, ce qui indique un **écart important** entre les valeurs réelles et la prédiction initiale.

---

### 🔹 Étape 3 : Calcul des résidus

Pour réduire cette perte, nous calculons les résidus, qui correspondent aux **erreurs résiduelles** entre la valeur réelle et la prédiction initiale :

| Étudiant | $y$  | $\hat{y}^{(0)}$ | $r_1$  |
|----------|------|----------------|------|
| A        | 10   | 14             | -4   |
| B        | 14   | 14             | 0    |
| C        | 18   | 14             | +4   |

Par conséquent, le **résidu** que nous calculons est directement :
$$
r_i^{(1)} = y_i - \hat{y}^{(0)}(x_i)
$$

> 🔎 Les résidus représentent la **correction nécessaire** pour réduire l'erreur globale. Ils sont utilisés comme **nouvelle cible** pour entraîner le prochain modèle faible.

---

🔎 **Légende** :  
- $y$ : note réelle  
- $\hat{y}^{(0)}$ : prédiction initiale (constante)  
- $r_1$ : résidu (erreur entre la valeur réelle et la prédiction initiale)  

---

### 📝 **Conclusion :**  
La fonction de perte utilisée ici est la **MSE**, dont le gradient est directement donné par les résidus. Le boosting consiste à entraîner un modèle faible pour corriger ces résidus, étape par étape, en réduisant ainsi progressivement l'erreur globale.  


### 🔹 Étape 4 : Construction du modèle faible $h_1(x)$

Le modèle faible est un **stump** (arbre de profondeur 1), c'est-à-dire qu'il fait **une seule coupure** pour **séparer les données**.

---

#### 💡 **1. Choix de la coupure optimale**

L'arbre de décision cherche la **meilleure coupure** qui **minimise l'erreur**.

- Les valeurs possibles pour couper sont **entre 2 et 4** ou **entre 4 et 6**.
- La **coupure optimale** est **entre B et C (4 et 6)**, car elle sépare les **résidus négatifs** (-4 et 0) des **résidus positifs** (+4).

---

#### 💡 **2. Pourquoi $h_1(x)$ est-il entraîné pour prédire les résidus ?**

- Dans Gradient Boosting, le modèle faible $h_1(x)$ est **entraîné pour prédire les résidus** car c'est **l'erreur restante** après la première prédiction.  
- Plutôt que de recalculer directement les prédictions, on cherche à **corriger progressivement l'erreur**.  
- Le modèle faible est donc **entraîné à apprendre la correction** à apporter pour **réduire l'erreur globale**.  
- En d'autres termes, **$h_1(x)$** est **ajusté** pour que sa sortie soit le **résidu prédit**.  

---

#### 💡 **3. Calcul de $h_1(x)$**  
Le modèle faible $h_1(x)$ prédit la **moyenne des résidus** dans chaque feuille, après avoir séparé les données en deux groupes :

**Groupe de gauche (A et B) :**

- Les résidus dans ce groupe sont : -4 et 0  
- La **moyenne des résidus** dans cette feuille est :  
  $$
  h_1(x) = \frac{-4 + 0}{2} = -2
  $$  
- Le modèle faible $h_1(x)$ prédit **-2** pour les étudiants A et B.

**Groupe de droite (C) :**

- Le résidu dans ce groupe est : +4  
- Puisqu'il n'y a qu'un seul élément, la **moyenne est simplement +4** :  
  $$
  h_1(x) = +4
  $$  
- Le modèle faible $h_1(x)$ prédit **+4** pour l'étudiant C.

---

#### 💡 **4. Prédictions du modèle faible $h_1(x)$**

| Étudiant | $x$ | $h_1(x)$ |
|---------|-----|---------|
| A       | 2   | -2      |
| B       | 4   | -2      |
| C       | 6   | +4      |

---

#### ✅ **Pourquoi -2, -2, +4 ?**
- Le modèle faible $h_1(x)$ est **entraîné sur les résidus** car ces résidus représentent les **erreurs à corriger** après la prédiction initiale.  
- Pour les étudiants A et B (groupe de gauche), la **moyenne des résidus** est **-2**.  
- Pour l'étudiant C (groupe de droite), la **moyenne des résidus** est **+4**.  
- Le **learning rate** pondère ensuite ces corrections pour éviter une mise à jour trop rapide.  

---

#### 💡 **5. Correction avec le learning rate $\gamma = 0.1$**

On applique le **learning rate** pour **réduire l'impact** des corrections :

- Prédiction pour A et B :  
  $$
  \text{Correction} = \gamma \times (-2) = 0.1 \times (-2) = -0.2
  $$  
- Prédiction pour C :  
  $$
  \text{Correction} = \gamma \times (+4) = 0.1 \times 4 = +0.4
  $$

---

#### 💡 **6. Nouvelles prédictions mises à jour**

| Étudiant | $\hat{y}^{(0)}$ | Correction ($h_1(x)$) | $\hat{y}^{(1)}$ (mise à jour) |
|---------|----------------|------------------------|-------------------------------|
| A       | 14             | -0.2                   | $14 + (-0.2) = 13.8$           |
| B       | 14             | -0.2                   | $14 + (-0.2) = 13.8$           |
| C       | 14             | +0.4                   | $14 + 0.4 = 14.4$              |

---

#### ✅ **Pourquoi cette mise à jour ?**
- Le modèle faible ($h_1(x)$) **prédit les résidus** pour corriger les erreurs de la prédiction initiale.  
- En ajoutant cette correction à la prédiction précédente, on **réduit progressivement l'erreur**.  
- L'ajustement est pondéré par le **learning rate** pour garantir une **correction progressive** sans surajustement.  
- Cette méthode permet de construire un modèle robuste en **accumulant de petites corrections** à chaque itération.  


## 🌟 Pourquoi parle-t-on de "prédiction" pendant l'entraînement ?

### 💡 Logique de la prédiction dans Gradient Boosting :
En Gradient Boosting, chaque modèle faible est **entraîné pour prédire les résidus** des prédictions précédentes.

1. **Modèle faible ($h_1(x)$) :**  
   - Le modèle faible est généralement un **arbre de décision simple**.  
   - Cet arbre est **entraîné spécifiquement pour prédire les erreurs** (résidus) commises par la prédiction précédente.  
   - On ne cherche pas à prédire directement la valeur cible ($y$), mais plutôt à **corriger l'erreur précédente**.  

2. **Pourquoi utiliser le mot "prédiction" ?**  
   - Même pendant l'entraînement, chaque modèle faible (comme un arbre) génère une **valeur estimée** (une prédiction) pour corriger l'erreur.  
   - Cette **prédiction des résidus** est ensuite ajoutée à la prédiction précédente pour mettre à jour la valeur finale.  

### 🔹 Étape 5 : Calcul de la nouvelle MSE après la première itération

Après la mise à jour avec le **modèle faible $h_1(x)$**, les nouvelles prédictions sont :

- A : 13.8  
- B : 13.8  
- C : 14.4  

Pour vérifier si le modèle a amélioré les prédictions, on **recalcule la MSE** après cette première itération.

---

#### 📏 **Nouvelle MSE :**

$$
\text{MSE}_{1} = \frac{1}{3} \left[ (10 - 13.8)^2 + (14 - 13.8)^2 + (18 - 14.4)^2 \right]
$$

#### 🧮 **Calcul détaillé :**

1. Pour l'étudiant A :  
   $$
   (10 - 13.8)^2 = (-3.8)^2 = 14.44
   $$

2. Pour l'étudiant B :  
   $$
   (14 - 13.8)^2 = (0.2)^2 = 0.04
   $$

3. Pour l'étudiant C :  
   $$
   (18 - 14.4)^2 = (3.6)^2 = 12.96
   $$

#### 💡 **Calcul final :**

$$
\text{MSE}_{1} = \frac{1}{3} \times (14.44 + 0.04 + 12.96) = \frac{27.44}{3} \approx 9.15
$$

---

#### 🔥 **Amélioration :**

- L'erreur a diminué de **10.67 à 9.15**, ce qui signifie que le **modèle faible a réussi à corriger une partie de l'erreur initiale**.  
- Cela montre que **l'ajout du modèle faible $h_1(x)$** a permis de **réduire la MSE**, ce qui est l'objectif de chaque itération dans le boosting.

## 2. 🔁 Deuxième itération

### 🔹 Étape 6 : Calcul des nouveaux résidus

On calcule les résidus après la première mise à jour des prédictions :

$$
r_i^{(2)} = y_i - \hat{y}^{(1)}(x_i)
$$

| Étudiant | $y$  | $\hat{y}^{(1)}$  | $r_2$   |
|----------|----|-----------------|-------|
| A        | 10 | 13.8            | -3.8  |
| B        | 14 | 13.8            | +0.2  |
| C        | 18 | 14.4            | +3.6  |

> 🔎 **Légende** :  
> - $y$ : note réelle  
> - $\hat{y}^{(1)}$ : prédiction après 1 itération  
> - $r_2$ : nouveau résidu (gradient)  

---

### 🔹 Étape 7 : Modèle 2 (entraînement sur les nouveaux résidus)

Supposons que le nouvel arbre (modèle 2) suive cette règle :

- Si $x < 3$ → prédire $-3.8$  
- Si $3 \leq x < 5$ → prédire $+0.2$  
- Si $x \geq 5$ → prédire $+3.6$  

| Étudiant | $x$  | $h_2(x)$ |
|----------|----|--------|
| A        | 2  | -3.8   |
| B        | 4  | +0.2   |
| C        | 6  | +3.6   |

> 🔎 $h_2(x)$ est le deuxième modèle faible entraîné sur les nouveaux résidus  

---

### 🔹 Étape 8 : Mise à jour de la prédiction

La mise à jour devient :

$$
\hat{y}^{(2)}(x) = \hat{y}^{(1)}(x) + 0.1 \cdot h_2(x)
$$

| Étudiant | $\hat{y}^{(2)}$ |
|----------|--------|
| A        | $13.8 + 0.1 \times (-3.8) = 13.42$ |
| B        | $13.8 + 0.1 \times (+0.2) = 13.82$ |
| C        | $14.4 + 0.1 \times (+3.6) = 14.76$ |

---

### ✅ **Pourquoi cette mise à jour ?**
- Le modèle faible ($h_2(x)$) est **entraîné pour prédire les nouveaux résidus** générés après la première itération.  
- La mise à jour est réalisée en appliquant un **learning rate ($\gamma = 0.1$)** pour **pondérer la correction**.  
- Les nouvelles prédictions sont légèrement ajustées pour progresser **par petites étapes** vers la convergence.

## 1. Données

|----------|--------------------------|-------------------|
| A        | 2                        | 10                |
| B        | 4                        | 14                |
| C        | 6                        | 18                |

---

## 2. Initialisation
Prédiction constante = moyenne des notes :

$$
\hat{y}^{(0)} = \frac{10+14+18}{3} = 14
$$

MSE initiale :

$$
\text{MSE}_0 = \frac{1}{3}[(10-14)^2+(14-14)^2+(18-14)^2] = 10.67
$$

---

## 3. Résidus de la première itération
$$
r^{(1)} = y - \hat{y}^{(0)}
$$

| Étudiant | $y$ | $\hat{y}^{(0)}$ | $r^{(1)}$ |
|----------|-----|-----------------|-----------|
| A        | 10  | 14              | -4        |
| B        | 14  | 14              | 0         |
| C        | 18  | 14              | +4        |

---

## 4. Modèle faible $h_1(x)$
Stump avec coupure entre B et C. Le choix de coupure dépend de MSE. 

- Groupe gauche (A,B) → moyenne = $(-4+0)/2 = -2$. ici MSE egal pour les deux possibilités de coupure, Il prévilège une séparation “négatif ou nul” vs “positif”
- Groupe droit (C) → moyenne = $+4$

| Étudiant | $h_1(x)$ |
|----------|----------|
| A        | -2       |
| B        | -2       |
| C        | +4       |

Mise à jour avec $\gamma=0.1$ :

$$
\hat{y}^{(1)} = \hat{y}^{(0)} + 0.1 \cdot h_1(x)
$$

| Étudiant | $\hat{y}^{(1)}$ |
|----------|-----------------|
| A        | 13.8            |
| B        | 13.8            |
| C        | 14.4            |

Nouvelle MSE ≈ 9.15 (amélioration).

---

## 5. Résidus de la deuxième itération
$$
r^{(2)} = y - \hat{y}^{(1)}
$$

| Étudiant | $y$ | $\hat{y}^{(1)}$ | $r^{(2)}$ |
|----------|-----|-----------------|-----------|
| A        | 10  | 13.8            | -3.8      |
| B        | 14  | 13.8            | +0.2      |
| C        | 18  | 14.4            | +3.6      |

---


## 6. Modèle faible $h_2(x)$

Stump avec coupure entre A et B :

- Groupe gauche (A) → moyenne = $-3.8$  
- Groupe droit (B,C) → moyenne = $(0.2+3.6)/2 = +1.9$

| Étudiant | $h_2(x)$ |
|----------|----------|
| A        | -3.8     |
| B        | +1.9     |
| C        | +1.9     |

Mise à jour avec $\gamma=0.1$ :

$$
\hat{y}^{(2)} = \hat{y}^{(1)} + 0.1 \cdot h_2(x)
$$

| Étudiant | $\hat{y}^{(2)}$ |
|----------|-----------------|
| A        | 13.42           |
| B        | 13.99           |
| C        | 14.59           |

---

### 📏 Calcul de la MSE pour la coupure A|B,C

Erreurs :
- $e_A = 10 - 13.42 = -3.42$  
- $e_B = 14 - 13.99 = +0.01$  
- $e_C = 18 - 14.59 = +3.41$

Carrés :
- $(-3.42)^2 \approx 11.70$  
- $(0.01)^2 \approx 0.0001$  
- $(3.41)^2 \approx 11.63$

MSE :
$$
\text{MSE}_2^{A|B,C} = \frac{11.70 + 0.0001 + 11.63}{3} \approx 7.78
$$

---

### 📏 Calcul de la MSE pour la coupure B|C

Stump alternatif :
- Groupe gauche (A,B) → moyenne = $(-3.8+0.2)/2 = -1.8$  
- Groupe droit (C) → $+3.6$

Corrections appliquées ($\gamma=0.1$) :
- A = $13.8 - 0.18 = 13.62$  
- B = $13.8 - 0.18 = 13.62$  
- C = $14.4 + 0.36 = 14.76$

Erreurs :
- $e_A = 10 - 13.62 = -3.62$  
- $e_B = 14 - 13.62 = +0.38$  
- $e_C = 18 - 14.76 = +3.24$

Carrés :
- $(-3.62)^2 \approx 13.10$  
- $(0.38)^2 \approx 0.1444$  
- $(3.24)^2 \approx 10.50$

MSE :
$$
\text{MSE}_2^{A,B|C} = \frac{13.10 + 0.1444 + 10.50}{3} \approx 7.91
$$

---

## ✅ Conclusion

- Chaque stump est entraîné sur les **résidus**.  
- Les corrections sont pondérées par le **learning rate**.  
- Les arbres ne corrigent jamais 100% des résidus, mais apportent des **moyennes de correction** par groupe.  
- La MSE diminue itération après itération → le modèle devient plus précis.  
- À l’itération 2, la coupure **A|B,C** est légèrement meilleure (MSE ≈ 7.78) que la coupure **A,B|C** (MSE ≈ 7.91).

### 🔹 Étape 2 : Pseudo-résidus (gradient de la log-loss)

XGBoost utilise comme "résidus" le **gradient négatif** de la fonction de perte log-loss :

$$
r_i^{(1)} = y_i - \hat{p}_i
$$

---

### 📊 Calcul des pseudo-résidus

| Étudiant | x (heures) | y (succès = 1 / échec = 0) | Probabilité prédite p₀ | Résidu r₁ = y - p₀ |
|----------|------------|-----------------------------|--------------------------|---------------------|
| A        | 2          | 0                           | 0.67                     | -0.67               |
| B        | 4          | 1                           | 0.67                     | +0.33               |
| C        | 6          | 1                           | 0.67                     | +0.33               |

---

🧠 Ces pseudo-résidus sont utilisés comme **cibles** pour entraîner un **premier arbre de décision**.


### 🔹 Étape 3 : Arbre 1 🌳 (entraîné sur les résidus)

Les pseudo-résidus calculés précédemment sont utilisés comme **cibles** pour entraîner un premier arbre de décision.

Supposons qu’on effectue une coupure à `x < 3` :

- A (x = 2) → -0.67  
- B, C (x = 4, 6) → +0.33

---

### 📈 Gain régularisé dans XGBoost

Le gain utilisé pour évaluer la qualité de ce split est :

$$
\text{Gain} = \frac{G^2}{H + \lambda}
$$

avec :

- **G** : somme des gradients dans le nœud  
- **H** : somme des hessiennes (approximations des 2ᵉ dérivées)  
- **$\lambda$** : paramètre de régularisation L2 (par défaut $\lambda = 1$=> Ridge ; $\lambda = 0$ => Lasso)

---

#### 💡 **Formule de calcul de la hessienne (H)**  
Dans le cas de la **classification binaire avec log loss**, la hessienne est calculée comme :  

$$
H = p \times (1 - p)
$$

où :  
- **$p$** : probabilité prédite  
- **$(1 - p)$** : complément de la probabilité  

---

#### 💡 **Étape 1 : Calcul des gradients (G)**  
Les gradients sont les **pseudo-résidus** calculés précédemment :  

| Étudiant | $x$ | $y$ | $p_0$ | $r_1 = y - p_0$ |
|---------|-----|-----|-------|-----------------|
| A       | 2   | 0   | 0.67  | -0.67            |
| B       | 4   | 1   | 0.67  | +0.33            |
| C       | 6   | 1   | 0.67  | +0.33            |

---

#### 💡 **Étape 2 : Calcul des sommes G et H**  

- **Nœud gauche (x < 3)**  
  - $G_{\text{gauche}} = -0.67$  
  - $H_{\text{gauche}} = 0.67 \times (1 - 0.67) = 0.22$  

- **Nœud droit (x \geq 3)**  
  - $G_{\text{droit}} = +0.33 + 0.33 = +0.66$  
  - $H_{\text{droit}} = 2 \times [0.67 \times (1 - 0.67)] = 2 \times 0.22 = 0.44$  

---

#### 💡 **Étape 3 : Calcul du gain**  
La formule du gain pour chaque nœud est :  

$$
\text{Gain} = \frac{G_{\text{gauche}}^2}{H_{\text{gauche}} + \lambda} + \frac{G_{\text{droit}}^2}{H_{\text{droit}} + \lambda} - \frac{(G_{\text{total}})^2}{H_{\text{total}} + \lambda}
$$  

- **Gain gauche :**  

$$
\frac{(-0.67)^2}{0.22 + 1} = \frac{0.4489}{1.22} = 0.368
$$  

- **Gain droit :**  

$$
\frac{(0.66)^2}{0.44 + 1} = \frac{0.4356}{1.44} = 0.303
$$  

- **Gain total :**  

$$
\frac{(0.33)^2 + (0.33)^2 + (-0.67)^2}{(0.22 + 0.22 + 0.22) + 1} = \frac{0.1089 + 0.1089 + 0.4489}{1.66} = \frac{0.6667}{1.66} = 0.402
$$  

---

#### ✅ **Gain final :**  

$$
\text{Gain} = (0.368 + 0.303) - 0.402 = 0.671 - 0.402 = 0.269
$$  

---

### 🌟 **Interprétation :**  
- Le **gain est positif (0.269)**, ce qui signifie que le **split améliore la modélisation**.  
- Si ce gain est **supérieur au seuil de régularisation (hyperparamètre gamma dans sklearn)**, le split est **conservé**.  
- Dans XGBoost, le split n'est **retenu que si le gain est significatif**, pour éviter le **surapprentissage**.

### 🔁 Mise à jour de la prédiction

On utilise ici un **learning rate** $\gamma = 0.1$.

La mise à jour de la prédiction suit la règle :

$$
F_1(x) = F_0(x) + \gamma \times h_1(x)
$$

---

### 🧮 Dans notre exemple :

- $F_0(x) \approx 0.69$ (log-odds initial pour $p = \frac{2}{3}$)  
- Les résidus sont :  
  - A : $-0.67$  
  - B, C : $+0.33$  

On entraîne un arbre **$h_1(x)$** qui fait le split suivant :

- Si $x < 3$ → prédire $-0.67$  
- Sinon → prédire $+0.33$  

| Étudiant | $x$ | $h_1(x)$ | $F_1(x) = F_0 + 0.1 \times h_1(x)$ |
|----------|---|---------|--------------------------|
| A        | 2 | -0.67   | $0.69 - 0.067 = 0.623$    |
| B        | 4 | +0.33   | $0.69 + 0.033 = 0.723$    |
| C        | 6 | +0.33   | $0.69 + 0.033 = 0.723$    |

---

### 🔄 **Probabilités mises à jour (sigmoïde)**

Pour obtenir les nouvelles probabilités, on applique la fonction sigmoïde :

$$
p_1(x) = \frac{1}{1 + \exp(-F_1(x))}
$$

| Étudiant | $F_1(x)$ | $p_1(x)$ (approx.) |
|----------|---------|-----------------|
| A        | 0.623   | $1 / (1 + \exp(-0.623)) \approx 0.65$ |
| B        | 0.723   | $1 / (1 + \exp(-0.723)) \approx 0.67$ |
| C        | 0.723   | $1 / (1 + \exp(-0.723)) \approx 0.67$ |

---

### 🔎 **Interprétation**

- Le modèle a **diminué la confiance** pour A (échec), et **renforcé** la probabilité de succès pour B et C.  
- L’arbre \( h_1(x) \) sert ici à **corriger** légèrement les prédictions de départ.  
- Cette étape se répétera à chaque itération avec un nouvel arbre, toujours sur les **résidus restants**.  

---

### 🧠 **En résumé :**

- **$F_1(x)$** est la nouvelle estimation “brute” (log-odds).  
- **$p_1(x)$** est la probabilité transformée via la fonction sigmoïde.  
- Le modèle progresse petit à petit vers une meilleure séparation entre classes en ajoutant chaque fois une **correction partielle**.  

---

### 🧠 **Spécificités visibles de XGBoost :**

- Choix de la coupure basé sur un **gain régularisé**.  
- Régularisation L2 limite les splits peu utiles.  
- **Early stopping** possible si la performance stagne.  
- Contrairement à AdaBoost ou Gradient Boosting, XGBoost **intègre ces mécanismes nativement**.

# 🌟 XGBoost – Classification en Deux Itérations

XGBoost améliore le Gradient Boosting avec :

- ✅ Une **régularisation explicite** (L1/L2) à chaque split  
- ✅ Un **système de gain régularisé** pour choisir les meilleures coupures  
- ✅ Une capacité d’**early stopping** pour éviter de surapprendre  

---

## 📊 1. Données

| Étudiant | $x$ (heures de révision) | $y$ (succès = 1 / échec = 0) |
|----------|--------------------------|------------------------------|
| A        | 2                        | 0                            |
| B        | 4                        | 1                            |
| C        | 6                        | 1                            |

---

## 🔁 2. Première itération

### 🔹 Étape 1 : Initialisation (log-odds)

$$
F_0(x) = \log\left(\frac{p}{1-p}\right), \quad p=\tfrac{2}{3} \Rightarrow F_0(x)=\log(2)\approx 0.693
$$

Probabilité initiale :

$$
\hat{p}_0(x) = \sigma(F_0(x)) = \frac{1}{1+e^{-0.693}} \approx 0.67
$$

---

### 🔹 Étape 2 : Pseudo-résidus

$$
r_i^{(1)} = y_i - \hat{p}_0(x_i)
$$

| Étudiant | $y$ | $\hat{p}_0$ | $r^{(1)}$ |
|----------|-----|-------------|-----------|
| A        | 0   | 0.67        | -0.67     |
| B        | 1   | 0.67        | +0.33     |
| C        | 1   | 0.67        | +0.33     |

---

### 🔹 Étape 3 : Construction de l’arbre $h_1(x)$

Critère de split (gain régularisé) :

$$
\text{Gain} = \tfrac{1}{2}\left[\frac{G_L^2}{H_L+\lambda} + \frac{G_R^2}{H_R+\lambda} - \frac{(G_L+G_R)^2}{H_L+H_R+\lambda}\right]
$$

Avec $\lambda=1$, hessienne $h=p(1-p)=0.22$ :

- **Split A | B,C** : Gain ≈ 0.27  
- **Split A,B | C** : Gain ≈ 0.08  

👉 Coupure optimale : **A | B,C**

Poids des feuilles :

$$
w_L=-\frac{G_L}{H_L+\lambda}\approx -0.55, \quad w_R=-\frac{G_R}{H_R+\lambda}\approx +0.46
$$

---

### 🔹 Étape 4 : Mise à jour des prédictions

Learning rate $\eta=0.1$ :

$$
F_1(x)=F_0(x)+\eta\cdot w_{\text{feuille}}
$$

| Étudiant | $h_1(x)$ | $F_1(x)$ | $p_1(x)=\sigma(F_1)$ |
|----------|----------|----------|-----------------------|
| A        | -0.55    | 0.693-0.055=0.638 | 0.65 |
| B        | +0.46    | 0.693+0.046=0.739 | 0.67 |
| C        | +0.46    | 0.739              | 0.67 |

---

## 🔁 3. Deuxième itération

### 🔹 Étape 1 : Nouveaux résidus

$$
r_i^{(2)} = y_i - p_1(x_i)
$$

| Étudiant | $y$ | $p_1(x)$ | $r^{(2)}$ |
|----------|-----|----------|-----------|
| A        | 0   | 0.65     | -0.65     |
| B        | 1   | 0.67     | +0.33     |
| C        | 1   | 0.67     | +0.33     |

---

### 🔹 Étape 2 : Construction de l’arbre $h_2(x)$

- **Split A | B,C** : Gain ≈ 0.32  
- **Split A,B | C** : Gain ≈ 0.08  

👉 Coupure optimale : **A | B,C**

Poids des feuilles :

$$
w_L\approx -0.53, \quad w_R\approx +0.45
$$

---

### 🔹 Étape 3 : Mise à jour des prédictions

$$
F_2(x)=F_1(x)+\eta\cdot w_{\text{feuille}}
$$

| Étudiant | $h_2(x)$ | $F_2(x)$ | $p_2(x)=\sigma(F_2)$ |
|----------|----------|----------|-----------------------|
| A        | -0.53    | 0.638-0.053=0.585 | 0.64 |
| B        | +0.45    | 0.739+0.045=0.784 | 0.69 |
| C        | +0.45    | 0.784              | 0.69 |

## ✅ Conclusion

- Chaque arbre est entraîné sur les **résidus** (gradients).  
- La **coupure optimale** est déterminée par le **gain régularisé**.  
- Les corrections sont pondérées par le **learning rate**.  
- Les probabilités évoluent progressivement vers les vraies classes :  
  - A (échec) → probabilité diminue vers 0  
  - B et C (succès) → probabilité augmente vers 1
