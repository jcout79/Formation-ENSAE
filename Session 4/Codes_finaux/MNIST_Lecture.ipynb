```python
import numpy as np

import matplotlib.pyplot as plt
```

```python
def load_mnist_images(filename):
    with open(filename, 'rb') as f:
        magic, num_images, rows, cols = np.frombuffer(f.read(16), dtype=np.uint32).byteswap().astype(np.int32)
        images = np.frombuffer(f.read(), dtype=np.uint8)
        return images.reshape(num_images, rows, cols) / 255.0

def load_mnist_labels(filename):
    with open(filename, 'rb') as f:
        magic, num_labels = np.frombuffer(f.read(8), dtype=np.uint32).byteswap().astype(np.int32)
        labels = np.frombuffer(f.read(), dtype=np.uint8)
        return labels
```
